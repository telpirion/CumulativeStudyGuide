## Correlation & regression fundamentals (review) — Questions 1–14

> **1.** \_\_\_\_\_\_\_ is a measure of the degree of relatedness of two variables.

* **Answer:** **`Correlation`**
* **Solution:**
  Correlation quantifies both the strength and direction of a linear relationship between two quantitative variables.

---

> **2.** The Pearson product-moment correlation coefficient is denoted by \_\_\_\_\_\_\_.

* **Answer:** **`r`**
* **Solution:**
  The sample Pearson product-moment correlation coefficient is denoted by $r$ (and the population correlation by $\rho$).

---

> **3.** The value of $r$ varies from \_\_\_\_\_\_\_.

* **Answer:** **`-1 to 0 to +1`**
* **Solution:**
  Pearson's $r$ is always bounded on the interval $[-1, +1]$.

---

> **4.** Perfect positive correlation results in an $r$ value of \_\_\_\_\_\_\_.

* **Answer:** **`+1`**
* **Solution:**
  When all $(x_i, y_i)$ observations lie exactly on a straight line with positive slope, $r = +1$.

---

> **5.** The value of the coefficient of correlation from the following data is \_\_\_\_\_\_\_.
>
> | $x$ | 19 | 20 | 26 | 31 | 34 | 45 | 45 | 51 |
> | :--- | :---: | :---: | :---: | :---: | :---: | :---: | :---: | :---: |
> | $y$ | 78 | 100 | 125 | 120 | 119 | 130 | 145 | 143 |

* **Answer:** **`0.876`** (`.876`)
* **Solution:**
  With $n = 8$, $\bar{x} = 33.875$, $\bar{y} = 120$, $SS_{xx} = 994.875$, $SS_{yy} = 3420$, and $SS_{xy} = 1616$:
  $$r = \frac{1616}{\sqrt{994.875 \times 3420}} \approx 0.876$$

---

> **6.** The value of $r$ from the following data is \_\_\_\_\_\_\_.
>
> | $x$ | -10 | -6 | 1 | 4 | 15 |
> | :--- | :---: | :---: | :---: | :---: | :---: |
> | $y$ | -26 | -44 | -36 | -39 | -43 |

* **Answer:** **`-0.581`** (`-.581`)
* **Solution:**

  $$r = \frac{-161.2}{\sqrt{374.8 \times 205.2}} \approx -0.581$$

---

> **7.** The process of constructing a mathematical model or function that can be used to predict or determine one variable by another variable is \_\_\_\_\_\_\_.

* **Answer:** **`Regression`**

> **8.** Bivariate linear regression is often termed \_\_\_\_\_\_\_ regression.

> **10.** In regression, the predictor is called the \_\_\_\_\_\_\_ variable.

## Regression line, residuals & standard error of the estimate — Questions 15–26

> **15.** In an effort to develop a regression model, the following data were gathered:
>
> | $x$ | 2 | 9 | 11 | 19 | 21 | 25 |
> | :--- | :---: | :---: | :---: | :---: | :---: | :---: |
> | $y$ | 26 | 17 | 18 | 15 | 15 | 8 |
>
> The slope of the regression line determined from these data is \_\_\_\_\_\_\_. The $y$ intercept is \_\_\_\_\_\_\_.

* **Answer:** **`-0.626`**, **`25.575`**
* **Solution:**
  With $n = 6$, $\bar{x} = 14.5$, $\bar{y} = 16.5$, $SS_{xx} = 371.5$, and $SS_{xy} = -232.5$:
  $$b_1 = \frac{-232.5}{371.5} \approx -0.626, \qquad b_0 = 16.5 - (-0.62584)(14.5) \approx 25.575$$

---

> **16.** A researcher wants to develop a regression line from the data given below:
>
> | $x$ | 12 | 11 | 5 | 6 | 9 |
> | :--- | :---: | :---: | :---: | :---: | :---: |
> | $y$ | 31 | 25 | 14 | 12 | 16 |
>
> The equation of the regression line is \_\_\_\_\_\_\_.

* **Answer:** **`ŷ = -1.253 + 2.425x`** (`-1.253 + 2.425 x`)
* **Solution:**
  With $n = 5$, $\bar{x} = 8.6$, $\bar{y} = 19.6$, $SS_{xx} = 37.2$, and $SS_{xy} = 90.2$:
  $$b_1 = \frac{90.2}{37.2} \approx 2.425, \qquad b_0 = 19.6 - 2.42473(8.6) \approx -1.253$$

---

> **17.** In regression, the value of $y - \hat{y}$ is called the \_\_\_\_\_\_\_.

* **Answer:** **`Residual`**
* **Solution:**
  The residual $e_i = y_i - \hat{y}_i$ measures the vertical error between an observed value and its predicted value.

---

> **18.** Data points that lie apart from the rest of the points are called \_\_\_\_\_\_\_.

* **Answer:** **`Outliers`**
* **Solution:**
  Outliers are observations that lie far outside the pattern formed by the rest of the sample.

---

> **19.** The regression assumption of constant error variance is called \_\_\_\_\_\_\_. If the error variances are not constant, it is called \_\_\_\_\_\_\_.

* **Answer:** **`Homoscedasticity`**, **`Heteroscedasticity`**
* **Solution:**
  Equal error variance across all levels of $X$ is homoscedasticity; changing (non-constant) error variance is heteroscedasticity.

---

> **20.** Suppose the following data are used to determine the equation of the regression line given below:
>
> | $x$ | 2 | 5 | 11 | 24 | 31 |
> | :--- | :---: | :---: | :---: | :---: | :---: |
> | $y$ | 12 | 13 | 16 | 14 | 19 |
>
> $$\hat{y} = 12.224 + 0.1764x$$
>
> The residual for $x = 11$ is \_\_\_\_\_\_\_.

* **Answer:** **`1.8356`**
* **Solution:**
  At $x = 11$, $y = 16$ and $\hat{y} = 12.224 + 0.1764(11) = 14.1644$:
  $$e = y - \hat{y} = 16 - 14.1644 = 1.8356$$

---

> **21.** The total of the residuals squared is called the \_\_\_\_\_\_\_.

* **Answer:** **`Sum of Squares of Error`** ($SSE$)
* **Solution:**
  $SSE = \sum_{i=1}^n (y_i - \hat{y}_i)^2$ measures unexplained residual variation.

---

> **22.** A standard deviation of the error of the regression model is called the \_\_\_\_\_\_\_ and is denoted by \_\_\_\_\_\_\_.

* **Answer:** **`Standard Error of the Estimate`**, **`s_e`**
* **Solution:**
  The standard error of the estimate ($s_e$) estimates the standard deviation of the residuals around the regression line.

---

> **23.** Suppose a regression model is developed for ten pairs of data resulting in $SSE = 1,203$. The standard error of the estimate is \_\_\_\_\_\_\_.

* **Answer:** **`12.263`**
* **Solution:**
  $$s_e = \sqrt{\frac{SSE}{n - 2}} = \sqrt{\frac{1203}{10 - 2}} = \sqrt{150.375} \approx 12.263$$

---

> **24.** A regression analysis results in the following data:
> $$\sum x = 276, \quad \sum x^2 = 12,014, \quad \sum xy = 2,438, \quad \sum y = 77, \quad \sum y^2 = 1,183, \quad n = 7$$
> The value of $SSE$ is \_\_\_\_\_\_\_.

* **Answer:** **`20.015`** (exact `20.016`)
* **Solution:**
  $$SS_{xx} = 12014 - \frac{276^2}{7} = 1131.7143, \quad SS_{xy} = 2438 - \frac{276(77)}{7} = -598$$
  $$b_1 = \frac{-598}{1131.7143} = -0.5284, \quad b_0 = 11 - (-0.5284)(39.4286) = 31.8341$$
  $$SSE = \sum y^2 - b_0 \sum y - b_1 \sum xy = 1183 - 31.8341(77) - (-0.5284)(2438) \approx 20.016$$

---

> **25.** The value of $s_e$ computed from the data of question 24 is \_\_\_\_\_\_\_.

* **Answer:** **`2.00`**
* **Solution:**
  $$s_e = \sqrt{\frac{SSE}{n - 2}} = \sqrt{\frac{20.016}{7 - 2}} = \sqrt{4.0032} \approx 2.00$$

---

> **26.** Suppose a regression model results in a value of $s_e = 27.9$. $95\%$ of the residuals should fall within \_\_\_\_\_\_\_.

* **Answer:** **`0 ± 55.8`**
* **Solution:**
  By the Empirical Rule for normally distributed residuals centered at $0$, approximately $95\%$ fall within $0 \pm 2s_e = 0 \pm 2(27.9) = 0 \pm 55.8$.

## Coefficient of determination, slope inference & trend line — Questions 27–36

> **27.** Coefficient of determination is denoted by \_\_\_\_\_\_\_.

* **Answer:** **`r^2`** ($r^2$ or $R^2$)
* **Solution:**
  In simple linear regression, the coefficient of determination equals $r^2$.

---

> **28.** \_\_\_\_\_\_\_ is the proportion of variability of the dependent variable accounted for or explained by the independent variable.

* **Answer:** **`Coefficient of Determination`**
* **Solution:**
  $r^2 = SSR / SST = 1 - SSE / SST$ represents the proportion of variation in $Y$ explained by $X$.

---

> **29.** The value of $r^2$ always falls between \_\_\_\_\_\_\_ and \_\_\_\_\_\_\_ inclusive.

* **Answer:** **`0`**, **`1`**
* **Solution:**
  Because $0 \le SSR \le SST$, $0 \le r^2 \le 1$.

---

> **30.** Suppose a regression analysis results in the following:
> $$b_1 = .19364, \quad b_0 = 59.4798, \quad n = 8, \quad \sum y = 1,019, \quad \sum y^2 = 134,451, \quad \sum xy = 378,932$$
> The value of $r^2$ for this regression model is \_\_\_\_\_\_\_.

* **Answer:** **`0.900`** (`.900`)
* **Solution:**
  $$SSE = 134451 - 59.4798(1019) - 0.19364(378932) \approx 464.71$$
  $$SST = 134451 - \frac{1019^2}{8} = 4655.875, \qquad r^2 = 1 - \frac{464.71}{4655.875} \approx 0.900$$

---

> **31.** Suppose the data below are used to determine the equation of a regression line:
>
> | $x$ | 18 | 14 | 9 | 6 | 2 |
> | :--- | :---: | :---: | :---: | :---: | :---: |
> | $y$ | 14 | 25 | 22 | 23 | 27 |
>
> The value of $r^2$ associated with this model is \_\_\_\_\_\_\_.

* **Answer:** **`0.578`** (`.578`)
* **Solution:**
  Fitting $\hat{y} = 28.0558 - 0.5975x$ gives $r = -0.7603$, so $r^2 = (-0.7603)^2 \approx 0.578$.

---

> **32.** A researcher has developed a regression model from sixteen pairs of data points. He wants to test to determine if the slope is significantly different from zero. He uses a two-tailed test and $\alpha = .01$. The critical table $t$ value is \_\_\_\_\_\_\_.

* **Answer:** **`± 2.977`** (`2.977`)
* **Solution:**
  With $n = 16$, $df = n - 2 = 14$. For a two-tailed test at $\alpha = 0.01$ ($\alpha / 2 = 0.005$), $t_{0.005, \, 14} = 2.977$.

---

> **33.** The following data are used to develop a simple regression model:
>
> | $x$ | 22 | 20 | 15 | 15 | 14 | 9 |

> The observed $t$ value used to test the slope of this regression model is \_\_\_\_\_\_\_.

* **Answer:** **`4.72`**

  $$t = \frac{1.8502}{0.3917} \approx 4.72$$

---

> **34.** If $\alpha = .05$ and a two-tailed test is being conducted, the critical table $t$ value to test the slope of the model developed in question 33 is \_\_\_\_\_\_\_.

* **Answer:** **`± 2.776`**
* **Solution:**
  With $n = 6$, $df = n - 2 = 4$. At $\alpha = 0.05$ (two-tailed, $\alpha / 2 = 0.025$), $t_{0.025, \, 4} = \pm 2.776$.

---

> **35.** The decision reached about the slope of the model computed in question 33 is to \_\_\_\_\_\_\_ the null hypothesis.

* **Answer:** **`Reject`**
* **Solution:**
  Since $|t_{\text{obs}}| = 4.72 > t_{\text{crit}} = 2.776$ ($p = 0.0092 < 0.05$), **reject** $H_0: \beta_1 = 0$.

# Chapter 13 study questions (Multiple Regression Analysis)

The following 7 study questions from `Chapter13PracticeProblems.pdf` review overall $F$-tests, individual partial $t$-tests, standard error of the estimate, $R^2$, Adjusted $R^2$, residual outlier detection, and interpreting computer regression output.

## Multiple regression evaluation & output interpretation — Questions 1–7

> **1.** In multiple regression, an \_\_\_\_\_\_\_ statistic is used to test for the overall effectiveness of the model.

* **Answer:** **`F`**
* **Solution:**
  The global $F$-test evaluates $H_0: \beta_1 = \beta_2 = \dots = \beta_k = 0$ against $H_a$: at least one $\beta_j \ne 0$.

---

> **2.** The significance of individual regression coefficients in a multiple regression model is tested using a \_\_\_\_\_\_\_ ratio.

* **Answer:** **`t`**
* **Solution:**
  Each partial slope $b_j$ is tested individually using $t = b_j / s_{b_j}$ with $df = n - k - 1$.

---

> **3.** The value, $s_e$, represents the \_\_\_\_\_\_\_.

* **Answer:** **`Standard Error of the Estimate`**
* **Solution:**
  In multiple regression with $k$ predictors, $s_e = \sqrt{SSE / (n - k - 1)}$ measures the standard deviation of the residuals around the fitted hyperplane.

---

> **4.** The coefficient of multiple determination is denoted by \_\_\_\_\_\_\_.

* **Answer:** **`R^2`** ($R^2$)
* **Solution:**
  $R^2 = SSR / SST = 1 - SSE / SST$ measures the proportion of total variation in $Y$ explained jointly by all $k$ predictors.

---

> **5.** Residuals can sometimes be used to locate \_\_\_\_\_\_\_ or values that are apart from the mainstream of the data.

* **Answer:** **`Outliers`**
* **Solution:**
  Observations with unusually large absolute or standardized residuals flag potential outliers or data-entry errors.

---

> **6.** Because $R^2$ may sometimes yield an inflated figure, statisticians have developed a(n) \_\_\_\_\_\_\_ to take into consideration both the additional information of each new independent variable and the changed degrees of freedom.

* **Answer:** **`Adjusted R^2`** ($R_{\text{adj}}^2$)
* **Solution:**
  $R_{\text{adj}}^2 = 1 - (1 - R^2)\frac{n - 1}{n - k - 1}$ penalizes adding predictors that do not sufficiently reduce $SSE$.

---

> **7.** Examine the computer output below taken from a multiple regression analysis with three independent variables.
>
> The regression equation is: $\hat{y} = 28.4 + 1.30x_1 - 0.25x_2 + 2.20x_3$
>
> | Predictor | Coef | Stdev | $t$-ratio | $p$ |
> | :--- | ---: | ---: | ---: | ---: |
> | Constant | 28.410 | 56.520 | 0.50 | 0.631 |
> | $x_1$ | 1.295 | 2.052 | 0.63 | 0.548 |
> | $x_2$ | -0.247 | 1.816 | -0.14 | 0.895 |
> | $x_3$ | 2.202 | 3.028 | 0.73 | 0.491 |
>
> *(Note: `x3` Stdev `3.028` appears clipped as `.028` in the PDF printout; $2.202 / 3.028 \approx 0.73$.)*
>
> $s_e = 32.68, \quad R^2 = 8.1\%, \quad R^2\text{ (adj)} = 0.0\%$
>
> **Analysis of Variance**
>
> | Source | $df$ | $SS$ | $MS$ | $F$ | $p$ |
> | :--- | ---: | ---: | ---: | ---: | ---: |
> | Regression | 3 | 659 | 220 | 0.21 | 0.889 |
> | Error | 7 | 7474 | 1068 | | |
> | Total | 10 | 8134 | | | |
>
> * The overall test of significance yields \_\_\_\_\_\_\_.
> * This test (is / is not) significant: \_\_\_\_\_\_\_.
> * The coefficient of multiple determination is \_\_\_\_\_\_\_.
> * The standard error of the estimate is \_\_\_\_\_\_\_.
> * The $t$-ratios and their associated probabilities indicate that \_\_\_\_\_\_\_ are significant predictor variables.
> * The regression coefficient of the $x_3$ variable is \_\_\_\_\_\_\_.
> * The value of the adjusted $R^2$ is \_\_\_\_\_\_\_.
> * This indicates \_\_\_\_\_\_\_.

* **Answers:**
  1. **`F = 0.21 with p = 0.889`**
  2. **`is not`** (`not`)
  3. **`0.081`** (`8.1%`)
  4. **`32.68`**
  5. **`None of the independent variables`**
  6. **`2.202`**
  7. **`0.000`** (`0.0%`, floored from negative raw adjusted $R^2$)
  8. **`There is virtually no predictability in this model`**
* **Solution:**
  With $n = 11$ ($df_{\text{Total}} = 10$) and $k = 3$:
  * $F = \frac{MSR}{MSE} = \frac{659 / 3}{7474 / 7} = \frac{219.67}{1067.71} \approx 0.21$ ($p = 0.889 > 0.05$), so the overall model is not significant.
  * Every partial $t$-ratio has $p \ge 0.491$, so none of $x_1, x_2, x_3$ are statistically significant.
  * $R^2 = 659 / 8134 = 0.0810$ ($8.1\%$), $s_e = \sqrt{7474 / 7} = 32.676 \approx 32.68$, and raw $R_{\text{adj}}^2 = 1 - (1 - 0.0810)\frac{10}{7} = -0.3128$, reported as $0.0\%$. 

```r
ss_reg <- 659
ss_err <- 7474
ss_tot <- 8134
df_reg <- 3
df_err <- 7
n_q7 <- df_reg + df_err + 1

ms_reg <- ss_reg / df_reg
ms_err <- ss_err / df_err
f_q7 <- ms_reg / ms_err
p_f_q7 <- pf(f_q7, df1 = df_reg, df2 = df_err, lower.tail = FALSE)
r2_q7 <- ss_reg / ss_tot
adj_r2_q7 <- 1 - ((1 - r2_q7) * (n_q7 - 1) / df_err)
se_q7 <- sqrt(ms_err)

print(sprintf("Q7 F = %.2f (p = %.3f) | s_e = %.2f | R^2 = %.3f | Raw Adj R^2 = %.4f (floored to 0.000)",
              f_q7, p_f_q7, se_q7, r2_q7, adj_r2_q7))
```

Compare results between the two models (`DIAM` is referred to as a covariate). Does its inclusion improve upon the fit of the first model without `DIAM`? Then evaluate the iterative $\log_{10}$ transformation (`log10(PRICE) ~ log10(DIAM) + TYPE`), Bartlett's test for homogeneity of variance, and robust regression (`MASS::rlm()`).
