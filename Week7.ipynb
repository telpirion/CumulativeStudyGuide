## Inferences about two means: independent samples ($\sigma$ known) — Questions 1–12

> **1.** A researcher wants to estimate the difference in the means of two populations. A random sample of 40 items from the first population results in a sample mean of 433 with a population standard deviation of 112. A random sample of 50 items from the second population results in a sample mean of 467 with a population standard deviation of 120. From this information, a point estimate of the difference of population means can be computed as \_\_\_\_\_\_\_.

* **Answer:** **`-34`**
* **Solution:**
  $$\text{Point Estimate} = \bar{x}_1 - \bar{x}_2 = 433 - 467 = -34$$

---

> **2.** Using the information from question 1, the researcher can compute a 95% confidence interval to estimate the difference in population means. The resulting confidence interval is \_\_\_\_\_\_\_.

* **Answer:** **`-82.07 <= µ1 - µ2 <= 14.07`** (or `[-82.07, 14.07]`)
* **Solution:**
  $$\sigma_{\bar{x}_1 - \bar{x}_2} = \sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}} = \sqrt{\frac{112^2}{40} + \frac{120^2}{50}} = \sqrt{313.6 + 288} = \sqrt{601.6} \approx 24.5275$$
  For 95% confidence, $z_{\alpha/2} = 1.96$:
  $$(\bar{x}_1 - \bar{x}_2) \pm z \cdot \sigma_{\bar{x}_1 - \bar{x}_2} = -34 \pm 1.96(24.5275) = -34 \pm 48.074 \implies [-82.07, 14.07]$$

---

> **3.** A random sample of 32 items is taken from a population which has a population variance of 93. The resulting sample mean is 45.6. A random sample of 37 items is taken from a population which has a population variance of 88. The resulting sample mean is 49.4. Using this information, a 98% confidence interval can be computed to estimate the difference in means of these two populations. The resulting interval is \_\_\_\_\_\_\_.

* **Answer:** **`-9.16 <= µ1 - µ2 <= 1.56`** (or `[-9.16, 1.56]`)
* **Solution:**
  $$\bar{x}_1 - \bar{x}_2 = 45.6 - 49.4 = -3.8$$
  $$\sigma_{\bar{x}_1 - \bar{x}_2} = \sqrt{\frac{93}{32} + \frac{88}{37}} = \sqrt{2.90625 + 2.37838} = \sqrt{5.28463} \approx 2.2988$$
  For 98% confidence ($\alpha = 0.02 \implies \alpha/2 = 0.01$), $z_{0.01} = 2.33$:
  $$-3.8 \pm 2.33(2.2988) = -3.8 \pm 5.356 \implies [-9.16, 1.56]$$

---

> **4.** A researcher desires to estimate the difference in means of two populations. To accomplish this, he/she takes a random sample of 85 items from the first population. The sample yields a mean of 168 with a population variance of 783. A random sample of 70 items is taken from the second population yielding a mean of 161 with a population variance of 780. A 94% confidence interval is computed to estimate the difference in population means. The resulting confidence interval is \_\_\_\_\_\_\_.

* **Answer:** **`-1.48 <= µ1 - µ2 <= 15.48`** (or `[-1.48, 15.48]`)
* **Solution:**
  $$\bar{x}_1 - \bar{x}_2 = 168 - 161 = 7$$
  $$\sigma_{\bar{x}_1 - \bar{x}_2} = \sqrt{\frac{783}{85} + \frac{780}{70}} = \sqrt{9.21176 + 11.14286} = \sqrt{20.3546} \approx 4.5116$$
  For 94% confidence ($\alpha = 0.06 \implies \alpha/2 = 0.03$), $z_{0.03} = 1.88$:
  $$7 \pm 1.88(4.5116) = 7 \pm 8.4818 \implies [-1.48, 15.48]$$

---

> **5.** Is there a difference in the average number years of experience of assembly line employees between company A and company B? A researcher wants to conduct a statistical test to answer this question. He is likely to be conducting a \_\_\_\_\_\_\_-tailed test.

* **Answer:** **`Two`**
* **Solution:** Because the research question asks whether "there is a difference" without asserting directionality, the hypotheses are $H_0: \mu_A - \mu_B = 0$ versus $H_a: \mu_A - \mu_B \ne 0$, necessitating a **two-tailed test**.

---

> **6.** The researcher who is conducting the test to determine if there is a difference in the average number of years of experience of assembly line workers between companies A and B is using an alpha of .10. The critical value of z for this problem is \_\_\_\_\_\_\_.

* **Answer:** **`± 1.645`** (or `1.645`)
* **Solution:** For a two-tailed test with $\alpha = 0.10$, the rejection area is split equally into $\alpha/2 = 0.05$ per tail. The critical values are $z = \pm 1.645$.

---

> **7.** Suppose the researcher conducting an experiment to compare the ages of workers at two companies. The researcher randomly samples forty-five assembly-line workers from company A and discovers that the sample average is 7.1 years with a population standard deviation of 2.3. Fifty-two assembly-line workers from company B are randomly selected resulting in a sample average of 6.2 years and a population standard deviation of 2.7. The observed z value for this problem is \_\_\_\_\_\_\_.

* **Answer:** **`1.77`**
* **Solution:**
  $$z = \frac{(\bar{x}_A - \bar{x}_B) - 0}{\sqrt{\frac{\sigma_A^2}{n_A} + \frac{\sigma_B^2}{n_B}}} = \frac{7.1 - 6.2}{\sqrt{\frac{2.3^2}{45} + \frac{2.7^2}{52}}} = \frac{0.9}{\sqrt{0.11756 + 0.14019}} = \frac{0.9}{\sqrt{0.25775}} = \frac{0.9}{0.50769} \approx 1.77$$

---

> **8.** Using an alpha of .10 and the critical values determined in questions 6 and 7, the decision is to \_\_\_\_\_\_\_ the null hypothesis.

* **Answer:** **`Reject`**
* **Solution:** Because observed $z = 1.77 > z_{\text{crit}} = 1.645$, the test statistic falls in the upper rejection region. The decision is to **reject** the null hypothesis.

---

> **9.** A researcher has a theory that the mean for population A is less than the mean for population B. To test this, she randomly samples thirty-eight items from population A and determines that the sample average is 38.4 with a population variance of 50.5 She randomly samples thirty-two items from population B and determines that the sample average is 44.3 with a population variance of 48.6 Alpha is .05. She is going to conduct a \_\_\_\_\_\_\_-tailed test.

* **Answer:** **`One`** (specifically, left-tailed)
* **Solution:** The alternative hypothesis specifies a strict direction ($H_a: \mu_A < \mu_B \iff \mu_A - \mu_B < 0$), making this a **one-tailed (left-tailed) test**.

---

> **10.** Using the information from question 9, the critical z value is \_\_\_\_\_\_\_.

* **Answer:** **`-1.645`**
* **Solution:** For a left-tailed test with $\alpha = 0.05$, the entire rejection region of area 0.05 is placed in the lower tail: $z_{\text{crit}} = -1.645$.

---

> **11.** Using the information from question 9, the observed value of z is \_\_\_\_\_\_\_.

* **Answer:** **`-3.50`**
* **Solution:**
  $$z = \frac{38.4 - 44.3}{\sqrt{\frac{50.5}{38} + \frac{48.6}{32}}} = \frac{-5.9}{\sqrt{1.32895 + 1.51875}} = \frac{-5.9}{\sqrt{2.8477}} = \frac{-5.9}{1.6875} \approx -3.50$$

---

> **12.** Using the results determined in question 10 and 11, the decision is to \_\_\_\_\_\_\_ the null hypothesis.

* **Answer:** **`Reject`**
* **Solution:** Because observed $z = -3.50 < z_{\text{crit}} = -1.645$, the test statistic falls far into the lower rejection region ($p < 0.0003$). The decision is to **reject** the null hypothesis.

## Inferences about two means: independent samples ($\sigma$ unknown) — Questions 13–19

> **13.** A researcher is interested in testing to determine if the mean of population one is greater than the mean of population two. He uses the following hypotheses to test this theory:
> $$H_0: \mu_1 - \mu_2 = 0$$
> $$H_a: \mu_1 - \mu_2 > 0$$
> He randomly selects a sample of 8 items from population one resulting in a mean of 14.7 and a standard deviation of 3.4. He randomly selects a sample of 12 items from population two resulting in a mean of 11.5 and a standard deviation 2.9. He is using an alpha value of .10 to conduct this test. The degrees of freedom for this problem are \_\_\_\_\_\_\_. It is assumed that these values are normally distributed in both populations.

* **Answer:** **`18`**
* **Solution:** For pooled-variance two-sample $t$-tests with assumed equal variances:
  $$df = n_1 + n_2 - 2 = 8 + 12 - 2 = 18$$

---

> **14.** The critical table t value used to conduct the hypothesis test in question 13 is \_\_\_\_\_\_\_.

* **Answer:** **`1.33`** (or `1.330`)
* **Solution:** For a right-tailed test ($H_a: \mu_1 - \mu_2 > 0$) with $\alpha = 0.10$ and $df = 18$, the critical value is $t_{0.10, 18} = 1.330$.

---

> **15.** The observed t value from the sample data is \_\_\_\_\_\_.

* **Answer:** **`2.26`**
* **Solution:**
  $$s_p^2 = \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2} = \frac{7(3.4)^2 + 11(2.9)^2}{18} = \frac{7(11.56) + 11(8.41)}{18} = \frac{173.43}{18} \approx 9.635$$
  $$t = \frac{(\bar{x}_1 - \bar{x}_2) - 0}{\sqrt{s_p^2 \left(\frac{1}{n_1} + \frac{1}{n_2}\right)}} = \frac{14.7 - 11.5}{\sqrt{9.635 \left(\frac{1}{8} + \frac{1}{12}\right)}} = \frac{3.2}{\sqrt{9.635(0.20833)}} = \frac{3.2}{\sqrt{2.00729}} = \frac{3.2}{1.4168} \approx 2.26$$

---

> **16.** Based on the observed t value obtained in question 15 and the critical table t value in question 14, the researcher should \_\_\_\_\_\_\_ the null hypothesis.

* **Answer:** **`Reject`**
* **Solution:** Because observed $t = 2.26 > t_{\text{crit}} = 1.33$, the test statistic falls in the upper rejection region ($p = 0.018$). The researcher should **reject** the null hypothesis.

---

## Inferences for related populations: matched pairs — Questions 20–33

> **20.** The matched-pairs t test deals with \_\_\_\_\_\_\_ samples.

* **Answer:** **`Related`** (or dependent)
* **Solution:** Matched-pairs designs evaluate paired or repeated measurements obtained from the same or matched subjects, representing **related** (dependent) samples.

> **21.** A researcher wants to conduct a before/after study on 13 subjects to determine if a treatment results in higher scores. The hypotheses are:
> $$H_0: D = 0$$
> $$H_a: D < 0$$
> Scores are obtained on the subjects both before and after the treatment. After subtracting the after scores from the before scores, the resulting value of $\bar{d}$ is -2.85 with a $s_d$ of 1.01. The degrees of freedom for this test are \_\_\_\_\_\_\_. Assume that the data are normally distributed in the population.

* **Answer:** **`12`**
* **Solution:** For paired difference tests, degrees of freedom are based on the number of pairs $n$:
  $$df = n - 1 = 13 - 1 = 12$$

---

> **22.** The critical table t value for the problem in question 21 is \_\_\_\_\_\_\_ if $\alpha = .01$.

* **Answer:** **`-2.681`**
* **Solution:** For a left-tailed test ($H_a: D < 0$) with $\alpha = 0.01$ and $df = 12$, the critical value is $-t_{0.01, 12} = -2.681$.

---

> **23.** The observed t value for the problem in question 21 is \_\_\_\_\_\_\_.

* **Answer:** **`-10.17`**
* **Solution:**
  $$t = \frac{\bar{d} - 0}{\frac{s_d}{\sqrt{n}}} = \frac{-2.85}{\frac{1.01}{\sqrt{13}}} = \frac{-2.85}{0.28012} \approx -10.17$$

---

> **24.** For the problem in question 21 based on the critical table t value obtained in question 22 and the observed t value obtained in question 23, the decision should be to \_\_\_\_\_\_\_ the null hypothesis.

* **Answer:** **`Reject`**
* **Solution:** Because observed $t = -10.17 < t_{\text{crit}} = -2.681$, the test statistic falls deeply into the lower rejection region. The decision is to **reject** the null hypothesis.

---

> **25.** A researcher is conducting a matched-pairs study. She gathers data on each pair in the study resulting in:

## Inferences about two population proportions — Questions 34–41

> **34.** A researcher is interested in estimating the difference in two populations proportions. A sample of 1000 from each population results in sample proportions of .61 and .64. The point estimate of the difference in the population proportions is \_\_\_\_\_\_\_.

* **Answer:** **`-.03`** (or `-0.03`)
* **Solution:** $\hat{p}_1 - \hat{p}_2 = 0.61 - 0.64 = -0.03$.

---

> **35.** Using the data from question 34, the researcher computes a 90% confidence interval to estimate the difference in population proportions. The resulting confidence interval is \_\_\_\_\_\_\_.

* **Answer:** **`-.066 <= p1 - p2 <= .006`** (or `[-0.066, 0.006]`)
* **Solution:**
  $$\text{SE} = \sqrt{\frac{p_1(1 - p_1)}{n_1} + \frac{p_2(1 - p_2)}{n_2}} = \sqrt{\frac{0.61(0.39)}{1000} + \frac{0.64(0.36)}{1000}} = \sqrt{0.0004683} \approx 0.02164$$
  For 90% confidence, $z_{\alpha/2} = 1.645$:
  $$-0.03 \pm 1.645(0.02164) = -0.03 \pm 0.0356 \implies [-0.066, 0.006]$$

---

> **36.** A random sample of 400 items from a population shows that 110 of the sample items possess a given characteristic. A random sample of 550 items from a second population resulted in 154 of the sample items possessing the characteristic. Using this data, a 99% confidence interval is constructed to estimate the difference in population proportions which possess the given characteristic. The resulting confidence interval is \_\_\_\_\_\_\_.

* **Answer:** **`-.081 <= p1 - p2 <= .071`** (or `[-0.081, 0.071]`)
* **Solution:**
  $$\hat{p}_1 = \frac{110}{400} = 0.275, \quad \hat{p}_2 = \frac{154}{550} = 0.280 \implies \hat{p}_1 - \hat{p}_2 = -0.005$$
  $$\text{SE} = \sqrt{\frac{0.275(0.725)}{400} + \frac{0.280(0.720)}{550}} = \sqrt{0.0004984 + 0.0003665} = \sqrt{0.000865} \approx 0.02941$$
  For 99% confidence, $z_{0.005} = 2.575$ (or $2.58$):
  $$-0.005 \pm 2.575(0.02941) = -0.005 \pm 0.0757 \implies [-0.081, 0.071]$$

---

> **37.** A researcher desires to estimate the difference in proportions of two populations. To accomplish this, he/she samples 338 and 332 items respectively from each population. The resulting sample proportions are .71 and .68 respectively. Using this data, a 90% confidence interval can be computed to estimate the difference in population proportions. The resulting confidence interval is \_\_\_\_\_\_\_.

* **Answer:** **`-.0285 <= p1 - p2 <= .0885`** (or `[-0.0285, 0.0885]`)
* **Solution:**
  $$\hat{p}_1 - \hat{p}_2 = 0.71 - 0.68 = 0.03$$
  $$\text{SE} = \sqrt{\frac{0.71(0.29)}{338} + \frac{0.68(0.32)}{332}} = \sqrt{0.0006092 + 0.0006554} = \sqrt{0.0012646} \approx 0.03556$$
  For 90% confidence, $z = 1.645$:
  $$0.03 \pm 1.645(0.03556) = 0.03 \pm 0.0585 \implies [-0.0285, 0.0885]$$

---

> **38.** A statistician is being asked to test a new theory that the proportion of population A possessing a given characteristic is greater than the proportion of population B possessing the characteristic. A random sample of 625 from population A has been taken and it is determined that 463 possess the characteristic. A random sample of 704 taken from population B results in 428 possessing the characteristic. The alternative hypothesis for this problem is \_\_\_\_\_\_\_.

* **Answer:** **`pA – pB > 0`** (or $H_a: p_A > p_B$)
* **Solution:** Testing whether proportion A is greater than proportion B translates to the directional alternative hypothesis $H_a: p_A - p_B > 0$.

---

> **39.** The observed value of z for question 38 is \_\_\_\_\_\_\_.

* **Answer:** **`5.14`**
* **Solution:**
  $$\hat{p}_A = \frac{463}{625} = 0.7408, \quad \hat{p}_B = \frac{428}{704} \approx 0.6080$$
  $$\bar{p} = \frac{463 + 428}{625 + 704} = \frac{891}{1329} \approx 0.67043, \quad \bar{q} = 1 - \bar{p} \approx 0.32957$$
  $$\text{SE}_{\text{pool}} = \sqrt{\bar{p}\bar{q}\left(\frac{1}{n_A} + \frac{1}{n_B}\right)} = \sqrt{(0.22096)\left(\frac{1}{625} + \frac{1}{704}\right)} = \sqrt{0.0006673} \approx 0.02583$$
  $$z = \frac{\hat{p}_A - \hat{p}_B}{\text{SE}_{\text{pool}}} = \frac{0.7408 - 0.6080}{0.02583} = \frac{0.1328}{0.02583} \approx 5.14$$

---

> **40.** Suppose alpha is .10. The critical value of z for question 38 is \_\_\_\_\_\_\_.

* **Answer:** **`1.28`**
* **Solution:** For a right-tailed test ($H_a: p_A - p_B > 0$) with $\alpha = 0.10$, the critical value is $z_{0.10} = 1.28$ (or $1.282$).

---

> **41.** Based on the results of question 39 and 40, the decision for the problem in question 38 is to \_\_\_\_\_\_\_ the null hypothesis.

* **Answer:** **`Reject`**
* **Solution:** Because observed $z = 5.14 > z_{\text{crit}} = 1.28$, the test statistic falls far into the upper rejection region. The decision is to **reject** the null hypothesis.

## Testing hypotheses about two population variances — Questions 42–44

> **42.** In testing hypotheses about two population variances, use the \_\_\_\_\_\_\_ distribution.

* **Answer:** **`F`**
* **Solution:** The sampling distribution of the ratio of two independent sample variances from normally distributed populations is the **$F$-distribution**.

---

> **43.** Suppose we want to test the following hypothesis:
> $$H_0: \sigma_1^2 = \sigma_2^2 \quad \text{and} \quad H_a: \sigma_1^2 > \sigma_2^2$$
> A sample of 9 items from population one yielded a sample standard deviation of 8.6. A sample of 8 items from population two yielded a sample standard deviation of 6.9. If alpha is .05, the critical F value is \_\_\_\_\_\_\_.

* **Answer:** **`3.73`**
* **Solution:**
  Numerator degrees of freedom: $df_1 = n_1 - 1 = 9 - 1 = 8$.
  Denominator degrees of freedom: $df_2 = n_2 - 1 = 8 - 1 = 7$.
  For a right-tailed test with $\alpha = 0.05$, $F_{0.05, 8, 7} = 3.73$.

> **44.** The observed F value for question 43 *(referenced as question 45 in original exam prompt)* is \_\_\_\_\_\_\_. The resulting decision is \_\_\_\_\_\_\_.

### Exercise 1

A double-blind clinical trial of a new drug for back pain was designed using control and treatment groups. Volunteers were fully informed and assigned at random to each group. Neither the volunteers nor the doctor knew when the new drug or a placebo was being administered. When 100 volunteers in each group had been treated and evaluated, the results revealed an 85% success rate for the new drug and a 65% success rate for the control group. At the 95% confidence level, is there a statistically significant difference between the two reported rates? Use a one-sided test. Also, report a confidence interval for the difference.

```r
# exercise-1
# Drug trial: Treatment (n = 100, x = 85), Control (n = 100, x = 65)
drug_x <- c(85, 65)
drug_n <- c(100, 100)

# One-sided test (alternative = "greater") at alpha = 0.05
prop_test_drug <- prop.test(drug_x, drug_n, alternative = "greater", conf.level = 0.95, correct = FALSE)
print(prop_test_drug)

# 95% confidence interval for the difference (p1 - p2)
ci_drug <- prop.test(drug_x, drug_n, conf.level = 0.95, correct = FALSE)$conf.int
print(ci_drug)
```

### Exercise 2

Two baseball players had their career records compared. In 267 times at bat, one player hit 85 home runs. In 248 times at bat, the other player hit 89 home runs. Assume the number of home runs follows a binomial distribution, is there a statistically significant difference with 95% confidence between the home run averages for these two baseball players?

```r
# exercise-2
# Player 1: 85 HR / 267 AB; Player 2: 89 HR / 248 AB
hr_x <- c(85, 89)
hr_n <- c(267, 248)

# Two-sided test at 95% confidence
prop_test_hr <- prop.test(hr_x, hr_n, alternative = "two.sided", conf.level = 0.95, correct = FALSE)
print(prop_test_hr)
```

### Exercise 3

Using the `home_prices` data (`data/home_prices.csv`), compare mean selling prices between homes located in the northeast sector of the city (`NBR`) versus the remaining homes. Also, compare the mean selling prices between homes with a corner lot (`CORNER`) and those located elsewhere. Use two-sample t-tests for the hypothesis tests at the 95% confidence level. Report confidence intervals for each.

```r
# exercise-3
# Compare PRICE by NBR (northeast sector vs remaining homes)
t_test_nbr <- t.test(PRICE ~ NBR, data = home_prices, conf.level = 0.95)
print(t_test_nbr)

# Compare PRICE by CORNER (corner lot vs elsewhere)
t_test_corner <- t.test(PRICE ~ CORNER, data = home_prices, conf.level = 0.95)
print(t_test_corner)
```

### Exercise 4

The `nsalary` data (`data/nsalary.csv`) are derived from data collected by the Department of Social Services of the State of New Mexico. The data have been adapted for this problem. Using these data compare mean salary levels between RURAL and non-RURAL locations. Produce a boxplot comparing RURAL and non-RURAL. Test equality of mean salaries using a two-sample t-test at the 95% confidence level. Report your results.

```r
# exercise-4
# Boxplot comparing RURAL and non-RURAL salaries
boxplot(NSAL ~ RURAL, data = nsalary,
        col = c("lightblue", "lightgreen"),
        main = "Annual Salary by Location",
        xlab = "Rural Location (NO / YES)",
        ylab = "Salary ($)")

# Two-sample t-test for equality of mean salaries at 95% confidence
t_test_salary <- t.test(NSAL ~ RURAL, data = nsalary, conf.level = 0.95)
print(t_test_salary)
```
