> **1.** Statistical measures used to yield information about the center or middle part of a group of numbers are called \_\_\_\_\_\_\_.

* **Answer:** **`Measures of Central Tendency`**
* **Solution:**
  Measures of central tendency quantify the center, typical value, or gravitational balance point of a numerical distribution (arithmetic mean, median, mode, percentiles, quartiles).


> **2.** The "average" is the \_\_\_\_\_\_\_.

* **Answer:** **`Mean`** (or Arithmetic Mean)
* **Solution:**
  In everyday parlance, "average" refers specifically to the arithmetic mean, calculated by summing all values and dividing by the total count.


> **3.** The value occurring most often in a group of numbers is called \_\_\_\_\_\_\_.

* **Answer:** **`Mode`**
* **Solution:**
  The mode is the data value associated with the greatest absolute frequency.


> **4.** In a set of 110 numbers arranged in order, the median is located at the \_\_\_\_\_\_\_ position.

* **Answer:** **`55.5th`**
* **Solution:**
  For an ordered array of size $n = 110$, the median position is $\frac{n + 1}{2} = \frac{110 + 1}{2} = 55.5$, which corresponds to the arithmetic average of the 55th and 56th values.


> **5.** If a set of data has an odd number of values arranged in ascending order, the median is the \_\_\_\_\_\_\_ value.

* **Answer:** **`Middle`** (at position $\frac{n+1}{2}$)
* **Solution:**
  When $n$ is odd, the middle observation splits the ordered distribution into two equal halves.


> **6.** Consider the data: `5, 4, 6, 6, 4, 5, 3, 2, 6, 4, 6, 3, 5`. The mode is \_\_\_\_\_\_\_.

* **Answer:** **`5`**

> **8.** Consider the data: `5, 4, 6, 6, 4, 5, 3, 2, 6, 4, 6, 3, 5`. The mean is \_\_\_\_\_\_\_.

> **9.** If a set of values is a population, then the mean is denoted by \_\_\_\_\_\_\_.

* **Answer:** **`μ`** (mu)
* **Solution:**

> **10.** In computing a mean for grouped data, the \_\_\_\_\_\_\_ is used to represent all data in a given class interval.

* **Answer:** **`Class Midpoint`** (Class Mark)
* **Solution:**
  Because individual raw data values within grouped intervals are unknown, the class midpoint $M_i = \frac{\text{Lower} + \text{Upper}}{2}$ is used as the proxy value for all observations in that class.

> **17.** Using the population data from Question 13, the value of the Pearsonian coefficient of skewness for these data is \_\_\_\_\_\_\_.

* **Answer:** **`-0.018`**
* **Solution:**
  $$\mu = \frac{908}{20} = 45.4, \quad \text{Median} = 45.5, \quad \sigma = \sqrt{\frac{\sum (x_i - 45.4)^2}{20}} = \sqrt{280.04} \approx 16.7344$$
$$S_k = \frac{3(\mu - \text{Median})}{\sigma} = \frac{3(45.4 - 45.5)}{16.7344} = \frac{-0.30}{16.7344} \approx -0.0179 \approx -0.018$$


> **18.** The Mean Absolute Deviation is computed by averaging the \_\_\_\_\_\_\_ of deviations around the mean.

* **Answer:** **`Absolute Value`**
* **Solution:**
  $$\text{MAD} = \frac{\sum |x_i - \bar{x}|}{n}$$


> **19.** Subtracting each value of a set of data from the mean produces \_\_\_\_\_\_\_ from the mean.

* **Answer:** **`Deviations`**
* **Solution:**
  The differences $(x_i - \mu)$ or $(x_i - \bar{x})$ represent deviations from the mean.


> **20.** The sum of the deviations from the mean is always \_\_\_\_\_\_\_.

* **Answer:** **`Zero`**
* **Solution:**
  $$\sum (x_i - \bar{x}) = \sum x_i - n\bar{x} = n\bar{x} - n\bar{x} = 0$$


> **21.** The variance is the \_\_\_\_\_\_\_ of the standard deviation.

* **Answer:** **`Square`**
* **Solution:**
  $$\sigma^2 = (\sigma)^2 \quad \text{and} \quad s^2 = (s)^2$$


> **22.** The population variance is computed by using \_\_\_\_\_\_\_ in the denominator, whereas the sample variance is computed by using \_\_\_\_\_\_\_ in the denominator.

* **Answer:** **`N, n - 1`**
* **Solution:**
  Sample variance divides by degrees of freedom ($n - 1$) rather than $n$ to provide an unbiased estimator of the population variance $\sigma^2$.


> **23.** If the sample standard deviation is 9, then the sample variance is \_\_\_\_\_\_\_.

* **Answer:** **`81`**
* **Solution:**
  $$s^2 = (9)^2 = 81$$


> **24.** Consider the data: `2, 3, 6, 12` ($n = 4$). The mean absolute deviation for this data is \_\_\_\_\_\_\_.

* **Answer:** **`3.25`**
* **Solution:**
  $$\bar{x} = \frac{2 + 3 + 6 + 12}{4} = 5.75$$
Deviations: $|2 - 5.75| = 3.75$, $|3 - 5.75| = 2.75$, $|6 - 5.75| = 0.25$, $|12 - 5.75| = 6.25$.
$$\text{MAD} = \frac{3.75 + 2.75 + 0.25 + 6.25}{4} = \frac{13.0}{4} = 3.25$$


> **25.** Consider the data: `2, 3, 6, 12` ($n = 4$). The sample variance for this data is \_\_\_\_\_\_\_.

* **Answer:** **`20.25`**
* **Solution:**
  Squared deviations: $(-3.75)^2 + (-2.75)^2 + (0.25)^2 + (6.25)^2 = 14.0625 + 7.5625 + 0.0625 + 39.0625 = 60.75$.
$$s^2 = \frac{60.75}{4 - 1} = \frac{60.75}{3} = 20.25$$


> **26.** Consider the data: `2, 3, 6, 12` ($N = 4$). The population standard deviation for this data is \_\_\_\_\_\_\_.

* **Answer:** **`3.897`**
* **Solution:**
  $$\sigma = \sqrt{\frac{60.75}{4}} = \sqrt{15.1875} \approx 3.8971 \approx 3.897$$


> **27.** In estimating what proportion of values fall within so many standard deviations of the mean, a researcher should use \_\_\_\_\_\_\_ if the shape of the distribution of numbers is unknown.

* **Answer:** **`Chebyshev’s Theorem`**
* **Solution:**
  Chebyshev's Theorem guarantees that at least $1 - \frac{1}{k^2}$ of data values fall within $\mu \pm k\sigma$ for any distribution shape where $k > 1$.


> **28.** Suppose a distribution of numbers is mound shaped with a mean of 150 and a variance of 225. Approximately \_\_\_\_\_\_\_ percent of the values fall between 120 and 180. Between \_\_\_\_\_\_\_ and \_\_\_\_\_\_\_ fall 99.7% of these values.

* **Answer:** **`95, 105, and 195`**
* **Solution:**
  $\sigma = \sqrt{225} = 15$. Under the Empirical Rule for mound-shaped distributions:
- $\mu \pm 2\sigma = 150 \pm 2(15) = [120, 180]$ contains approx. **95%**.
- $\mu \pm 3\sigma = 150 \pm 3(15) = [105, 195]$ contains approx. **99.7%**.


> **29.** The shape of a distribution of numbers is unknown. The distribution has a mean of 275 and a standard deviation of 12. The value of $k$ for 299 is \_\_\_\_\_\_\_. At least \_\_\_\_\_\_\_ percent of the values fall between 251 and 299.

* **Answer:** **`2, 75`**
* **Solution:**
  $$k = \frac{299 - 275}{12} = \frac{24}{12} = 2$$
By Chebyshev's Theorem, at least $1 - \frac{1}{2^2} = 1 - 0.25 = 75\%$ of values lie within $275 \pm 2(12) = [251, 299]$.


> **30.** Suppose data are normally distributed with a mean of 36 and a standard deviation of 4.8. The z score for 30 is \_\_\_\_\_\_\_. The z score for 40 is \_\_\_\_\_\_\_.

$$z_{40} = \frac{40 - 36}{4.8} = \frac{4}{4.8} \approx 0.8333 \approx 0.83$$


> **31.** A normal distribution of values has a mean of 74 and a standard deviation of 21. The coefficient of variation for this distribution is \_\_\_\_\_\_\_.

* **Answer:** **`28.38%`**
* **Solution:**
  $$CV = \left(\frac{\sigma}{\mu}\right) \times 100\% = \left(\frac{21}{74}\right) \times 100\% \approx 28.3784\% \approx 28.38\%$$

### Exercise 1: EPA Passenger Car Mileage Analysis

**Dataset:** `data/mileage.csv` is derived from a 1991 U.S. EPA study of passenger car mileage. This file includes information on sixty cars: `HP` (engine horsepower), `MPG` (average miles per gallon), `WT` (vehicle weight in 100 lb units), and `CLASS` (vehicle weight class C1 through C6).

1. For each weight class determine the mean and standard deviation of `MPG`. What can you conclude from these calculations?
2. For each weight class determine the mean and standard deviation of `HP`. What can you conclude from these calculations?

```r
mpg_mean <- aggregate(MPG ~ CLASS, data = mileage, mean)
mpg_sd <- aggregate(MPG ~ CLASS, data = mileage, sd)
mpg_summary <- data.frame(CLASS = mpg_mean$CLASS, Mean_MPG = round(mpg_mean$MPG, 2), SD_MPG = round(mpg_sd$MPG, 2))
print(mpg_summary)

hp_mean <- aggregate(HP ~ CLASS, data = mileage, mean)
hp_sd <- aggregate(HP ~ CLASS, data = mileage, sd)
hp_summary <- data.frame(CLASS = hp_mean$CLASS, Mean_HP = round(hp_mean$HP, 2), SD_HP = round(hp_sd$HP, 2))
print(hp_summary)
```

**Conclusions for Exercise 1:**

* **MPG by Weight Class:** There is low variability (standard deviation $\le 2.70$ mpg) within each vehicle weight class, but large, systematic differences between weight classes. Mean fuel economy drops monotonically as vehicle weight increases—from $39.68$ MPG in the lightest class (C1) down to $19.19$ MPG in the heaviest class (C6). This confirms a strong inverse relationship between vehicle weight and fuel efficiency.
* **Horsepower (HP) by Weight Class:** Average engine horsepower increases progressively with vehicle weight class (from $89.22$ HP in C1 to $224.71$ HP in C6). Crucially, the dispersion of horsepower expands dramatically: standard deviation is modest for lighter cars ($SD = 7.05$ HP in C1 and $9.09$ HP in C2), but expands substantially in heavier categories ($SD = 45.35$ HP in C5 and $74.02$ HP in C6), reflecting greater engine diversity among luxury and high-performance heavy vehicles.

### Exercise 2: Store Shopper Dollar Spending Distribution

**Dataset:** `data/shoppers.csv` contains the dollar amounts spent in a store by individual shoppers during one day.

1. Find the mean, median, range, standard deviation, variance, $Q_1$, $Q_3$, and $P_{10}$.
2. Plot the histogram and describe the distribution.

```r
calc_range <- function(x) { max(x, na.rm = TRUE) - min(x, na.rm = TRUE) }
summary_stats <- function(x) {
  stats <- data.frame(
    Value = c(
      mean(x, na.rm = TRUE),
      median(x, na.rm = TRUE),
      calc_range(x),
      sd(x, na.rm = TRUE),
      var(x, na.rm = TRUE),
      quantile(x, probs = 0.25, na.rm = TRUE),
      quantile(x, probs = 0.75, na.rm = TRUE),
      quantile(x, probs = 0.10, na.rm = TRUE)
    ),
    row.names = c("Mean", "Median", "Range", "StdDev", "Var", "Q1", "Q3", "P10")
  )
  round(stats, 4)
}
print(summary_stats(shoppers$Spending))

hist(shoppers$Spending, col = "steelblue", xlab = "Spending ($)",
     main = "Histogram of Shopper Spending")
```

**Description of Shopper Spending Distribution:**

- The distribution of customer spending is unimodal and **distinctly positively skewed (right-skewed)**.
- The arithmetic mean ($\bar{x} = \$25.44$) is noticeably higher than the median ($\text{Median} = \$20.74$), pulled upward by a small proportion of high-value shoppers.
- The middle 50% of customers spend between $Q_1 = \$14.39$ and $Q_3 = \$33.67$ ($IQR = \$19.27$), while 10% of customers spend $\le \$10.18$ ($P_{10}$). The total spending range spans $\$61.53$ (from $\$2.32$ to $\$63.85$).

**Conclusions for Exercise 3:**

* **Why Difference of Averages $\ne$ Average of Pairwise Differences:**
  - In part (2), the average President height is computed across all $n = 38$ administrations ($\bar{x} = 179.6842$ cm), while the opponent average is computed across the $n = 34$ non-missing opponents ($\bar{y} = 179.9706$ cm). The difference of these two group means is $179.6842 - 179.9706 = -0.2864$ cm.
  - In part (3), pairwise differences can only be computed for elections where *both* candidates have recorded heights ($n = 34$, excluding early uncontested elections such as Washington and Monroe). For those 34 contested elections, the mean president height is $180.0294$ cm and the mean opponent height is $179.9706$ cm, producing a mean pairwise difference of $180.0294 - 179.9706 = +0.0588$ cm.
  - The numerical discrepancy occurs entirely because the two calculations operate on different underlying sample subsets due to missing opponent data in uncontested elections.

### Exercise 4: Old Faithful Geyser Eruption Intervals

**Dataset:** `data/geyser.csv` contains the intervals (in minutes) between eruptions of Old Faithful Geyser in Yellowstone National Park taken on two consecutive weeks (`WEEK1` and `WEEK2`).

Compare the two sets of data using `summary()`, `hist()`, and `boxplot()`. What do you conclude?

```r
geyser_summary <- apply(geyser, 2, summary)
print(geyser_summary)

hist(geyser$WEEK1, col = "lightcyan", xlab = "Interval (Minutes)",
     main = "Histogram of Old Faithful Intervals: Week 1")

hist(geyser$WEEK2, col = "mistyrose", xlab = "Interval (Minutes)",
     main = "Histogram of Old Faithful Intervals: Week 2")

boxplot(geyser$WEEK1, geyser$WEEK2, names = c("Week 1", "Week 2"),
        horizontal = TRUE, col = c("lightcyan", "mistyrose"),
        xlab = "Interval (Minutes)", main = "Boxplots of Eruption Intervals by Week")
```

**Conclusions for Exercise 4:**

* **Central Tendency:** Eruption intervals across both weeks display remarkably stable central values:
  - Week 1: $\text{Mean} = 71.62$ minutes, $\text{Median} = 75.00$ minutes
  - Week 2: $\text{Mean} = 72.76$ minutes, $\text{Median} = 76.00$ minutes

* **Summary:** The geyser demonstrates consistent cyclical geothermal behavior between consecutive weeks, governed by identical underlying recharge mechanisms that alternate between short and long intervals.

### Exercise 6: Outlier Detection, Masking, and Winsorized Variance (Wilcox Ch. 2)

**Benchmark Source:** Rand Wilcox (*Basic Statistics*, Chapter 2, pp. 23–28). Consider the dataset containing eight small observations and two extreme outliers:
`wilcox_data = c(2, 2, 3, 3, 3, 4, 4, 4, 100000, 100000)`

1. Apply the classical outlier detection rule: declare an observation $x$ an outlier if $\frac{|x - \bar{x}|}{s} \ge 2$. Does the classical rule identify the extreme values $100,000$?
2. Explain why the classical rule suffers from **masking**.
3. Apply Tukey's Boxplot Rule (via `boxplot.stats()`). Does the boxplot rule detect the outliers?
4. Calculate the 20% Winsorized sample variance ($s_w^2$) and compare it with the classical sample variance ($s^2$).

```r
w_data <- c(2, 2, 3, 3, 3, 4, 4, 4, 100000, 100000)
n_w <- length(w_data)
mean_w <- mean(w_data)
sd_w <- sd(w_data)
var_w <- var(w_data)

z_stats <- abs(w_data - mean_w) / sd_w
max_z <- max(z_stats)

bp_w <- boxplot.stats(w_data)
lower_hinge <- bp_w$stats[2]
upper_hinge <- bp_w$stats[4]
iqr_h <- upper_hinge - lower_hinge
upper_fence <- upper_hinge + 1.5 * iqr_h

g_w <- floor(0.2 * n_w)
winsor_vec <- sort(w_data)
winsor_vec[1:g_w] <- winsor_vec[g_w + 1]
winsor_vec[(n_w - g_w + 1):n_w] <- winsor_vec[n_w - g_w]
winsor_mean <- mean(winsor_vec)
winsor_var <- var(winsor_vec)

masking_results <- data.frame(
  Metric = c("Classical Mean", "Classical StdDev", "Max Classical |z|",
             "Classical Outliers Flagged (|z| >= 2)", "Tukey Upper Fence",
             "Boxplot Outliers Detected", "Classical Sample Variance (s^2)",
             "20% Winsorized Variance (s_w^2)"),
  Value = c(round(c(mean_w, sd_w, max_z), 2),
            "0 (Failed / Masked)",
            round(upper_fence, 2),
            "100000, 100000 (Successful)",
            round(var_w, 2),
            round(winsor_var, 4))
)
print(masking_results)
```

**Conclusions for Exercise 6:**

* **Masking in Classical Outlier Detection:** For the observations `2, 2, 3, 3, 3, 4, 4, 4, 100000, 100000`, the sample mean is $\bar{x} = 20,002.50$ and the standard deviation is $s = 42,162.38$. Computing the test statistic yields $\frac{|100,000 - 20,002.50|}{42,162.38} = 1.897 < 2$. Thus, the classical rule fails to declare $100,000$ an outlier! The presence of the two massive outliers so grossly inflates the sample standard deviation that it masks both outliers from detection.
* **Superiority of the Boxplot Rule:** Because quartiles $Q_1 = 3$ and $Q_3 = 4$ depend only on rank order, the interquartile range remains unaffected by extreme magnitude ($IQR = 1$). The upper fence is $4 + 1.5(1) = 5.5$, cleanly declaring both $100,000$ observations as outliers.
* **Stability of Winsorized Variance:** While the classical sample variance explodes to $s^2 = 1,777,666,669$, the 20% Winsorized variance replaces the extreme tails with adjacent values ($3$ and $4$), yielding $s_w^2 = 0.2778$. Winsorized variance reflects the true dispersion of the non-contaminated data and remains stable against tail contamination.
