## Continuous Uniform Distribution — Questions 1–4

> **1.** The uniform distribution is sometimes referred to as the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ distribution.

* **Answer:** **`Rectangular`**
* **Solution:**
  Because its probability density function $f(x) = \frac{1}{b - a}$ has constant height across the interval $[a, b]$, the continuous uniform graph forms a rectangle.

---

> **2.** Suppose a set of data are uniformly distributed from $x = 5$ to $x = 13$. The height of the distribution is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_. The mean of this distribution is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_. The standard deviation of this distribution is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`1/8` (`0.125`), `9`, `2.3094`**
* **Solution:**
  With $a = 5$ and $b = 13$:
  * Height: $f(x) = \frac{1}{b - a} = \frac{1}{13 - 5} = \frac{1}{8} = 0.125$
  * Mean: $\mu = \frac{a + b}{2} = \frac{5 + 13}{2} = 9$
  * Standard Deviation: $\sigma = \frac{b - a}{\sqrt{12}} = \frac{13 - 5}{\sqrt{12}} = \frac{8}{3.464102} \approx 2.3094$

---

> **3.** Suppose a set of data are uniformly distributed from $x = 27$ to $x = 44$. The height of this distribution is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_. The mean of this distribution is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_. The standard deviation of this distribution is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`1/17` (`0.0588`), `35.5`, `4.9075`**
* **Solution:**
  With $a = 27$ and $b = 44$:
  * Height: $f(x) = \frac{1}{44 - 27} = \frac{1}{17} \approx 0.0588$
  * Mean: $\mu = \frac{27 + 44}{2} = 35.5$
  * Standard Deviation: $\sigma = \frac{44 - 27}{\sqrt{12}} = \frac{17}{\sqrt{12}} \approx 4.9075$

---

> **4.** A set of values is uniformly distributed from $84$ to $98$. The probability of a value occurring between $89$ and $93$ is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_. The probability of a value occurring between $80$ and $90$ is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_. The probability of a value occurring that is greater than $75$ is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`0.2857`, `0.4286` *(textbook PDF key prints `.7143`)*, `1.000`**
* **Solution:**
  With $a = 84$ and $b = 98$ (width $b - a = 14$):
  * $P(89 \le X \le 93) = \frac{93 - 89}{98 - 84} = \frac{4}{14} \approx 0.2857$
  * $P(80 \le X \le 90) = P(84 \le X \le 90) = \frac{90 - 84}{98 - 84} = \frac{6}{14} \approx 0.4286$. *(Note: The source PDF answer key lists `.7143`, which mistakenly computes $\frac{90 - 80}{14}$ instead of clamping the lower bound to the support minimum $a = 84$.)*
  * $P(X > 75) = P(84 \le X \le 98) = \frac{98 - 84}{98 - 84} = 1.0000$, because the entire support $[84, 98]$ lies above $75$.

```r
print(sprintf("Q2 -> height = %.4f | mean = %.2f | sd = %.4f", 1 / (13 - 5), (5 + 13) / 2, (13 - 5) / sqrt(12)))
print(sprintf("Q3 -> height = %.4f | mean = %.2f | sd = %.4f", 1 / (44 - 27), (27 + 44) / 2, (44 - 27) / sqrt(12)))
print(sprintf("Q4 -> P(89<=X<=93) = %.4f | P(80<=X<=90) = %.4f | P(X>75) = %.4f",
              punif(93, 84, 98) - punif(89, 84, 98),
              punif(90, 84, 98) - punif(80, 84, 98),
              punif(75, 84, 98, lower.tail = FALSE)))
```

## Normal Distribution & $z$-Scores — Questions 5–13

> **5.** Probably the most widely known and used of all distributions is the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ distribution.

* **Answer:** **`Normal`**
* **Solution:**
  The normal (Gaussian) distribution is foundational across inferential statistics, natural sciences, and business analytics.

---

> **6.** Many human characteristics can be described by the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ distribution.

* **Answer:** **`Normal`**
* **Solution:**
  Height, weight, IQ scores, and many biological/measurement characteristics follow an approximately normal distribution.

---

> **7.** The area under the curve of a normal distribution is \_\_\_\_\_\_\_\_.

* **Answer:** **`1`**
* **Solution:**
  All valid probability density functions integrate to a total area of $1.00$.

---

> **8.** In working normal curve problems using the raw values of $x$, the mean, and the standard deviation, a problem can be converted to \_\_\_\_\_\_\_\_ scores.

* **Answer:** **`z`**
* **Solution:**
  Standardizing raw values via $z = \frac{x - \mu}{\sigma}$ transforms any normal distribution $N(\mu, \sigma^2)$ into the standard normal distribution $Z \sim N(0, 1)$.

---

> **9.** A $z$ score value is the number of \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ a value is from the mean.

* **Answer:** **`Standard deviations`**
* **Solution:**
  A $z$-score measures the signed distance between an observation $x$ and the mean $\mu$ in units of $\sigma$.

---

> **10.** Within a range of $z$ scores of $\pm 1\sigma$ from the mean, fall \_\_\_\_\_\_\_\_$\%$ of the values of a normal distribution.

* **Answer:** **`68%`** ($68.26\%$)
* **Solution:**
  By the Empirical Rule, $P(-1 \le Z \le 1) \approx 0.6826$, or approximately $68\%$.

---

> **11.** Suppose a population of values is normally distributed with a mean of $155$ and a standard deviation of $12$. The $z$ score for $x = 170$ is \_\_\_\_\_\_\_\_.

* **Answer:** **`1.25`**
* **Solution:**
  $$z = \frac{170 - 155}{12} = \frac{15}{12} = 1.25$$

---

> **12.** Suppose a population of values is normally distributed with a mean of $76$ and a standard deviation of $5.2$. The $z$ score for $x = 73$ is \_\_\_\_\_\_\_\_.

* **Answer:** **`-0.58`** (`-0.5769` exact)
* **Solution:**
  $$z = \frac{73 - 76}{5.2} = \frac{-3}{5.2} \approx -0.5769 \approx -0.58$$

---

> **13.** Suppose a population of values is normally distributed with a mean of $250$ and a variance of $225$. The $z$ score for $x = 286$ is \_\_\_\_\_\_\_\_.

* **Answer:** **`2.40`**
* **Solution:**
  Since $\sigma^2 = 225$, the standard deviation is $\sigma = \sqrt{225} = 15$:
  $$z = \frac{286 - 250}{15} = \frac{36}{15} = 2.40$$

## Normal Probabilities & Inverse Quantiles — Questions 14–21

> **14.** Suppose a population of values is normally distributed with a mean of $9.8$ and a standard deviation of $2.5$. The probability that a value is greater than $11$ in the distribution is \_\_\_\_\_\_\_\_.

* **Answer:** **`0.3156`**
* **Solution:**
  $$z = \frac{11 - 9.8}{2.5} = 0.48 \implies P(X > 11) = P(Z > 0.48) = 0.3156$$

---

> **15.** A population is normally distributed with a mean of $80$ and a variance of $400$. The probability that $x$ lies between $50$ and $100$ is \_\_\_\_\_\_\_\_.

* **Answer:** **`0.7745`**
* **Solution:**
  With $\sigma = \sqrt{400} = 20$:
  $$z_1 = \frac{50 - 80}{20} = -1.50, \qquad z_2 = \frac{100 - 80}{20} = +1.00$$
  $$P(50 \le X \le 100) = P(-1.50 \le Z \le 1.00) = 0.8413 - 0.0668 = 0.7745$$

---

> **16.** A population is normally distributed with a mean of $115$ and a standard deviation of $13$. The probability that a value is less than $85$ is \_\_\_\_\_\_\_\_.

* **Answer:** **`0.0104`** *(Z-table rounded $z = -2.31$)*; **`0.0105`** *(exact in R)*
* **Solution:**
  $$z = \frac{85 - 115}{13} \approx -2.3077 \approx -2.31 \implies P(Z < -2.31) = 0.0104$$
  In R without intermediate rounding, `pnorm(85, 115, 13) = 0.010508`.

---

> **17.** A population is normally distributed with a mean of $64$. The probability that a value from this population is more than $70$ is $0.0485$. The standard deviation is \_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`3.614`** (`3.6154` exact in R)
* **Solution:**
  The upper tail probability $0.0485$ corresponds to cumulative probability $1 - 0.0485 = 0.9515$, giving $z = 1.66$:
  $$1.66 = \frac{70 - 64}{\sigma} \implies \sigma = \frac{6}{1.66} \approx 3.6145$$

---

> **18.** A population is normally distributed with a mean of $90$. $85.99\%$ of the values in this population are greater than $75$. The standard deviation of this population is \_\_\_\_\_\_\_\_.

* **Answer:** **`13.89`**
* **Solution:**
  $P(X > 75) = 0.8599 \implies P(X \le 75) = 0.1401$, which gives $z = -1.08$:
  $$-1.08 = \frac{75 - 90}{\sigma} \implies \sigma = \frac{-15}{-1.08} \approx 13.89$$

---

> **19.** A population is normally distributed with a standard deviation of $18.5$. $69.85\%$ of the values in this population are greater than $93$. The mean of the population is \_\_\_\_\_\_\_\_.

* **Answer:** **`102.62`**
* **Solution:**
  $P(X > 93) = 0.6985 \implies P(X \le 93) = 0.3015$, which gives $z = -0.52$:
  $$-0.52 = \frac{93 - \mu}{18.5} \implies \mu = 93 - (-0.52)(18.5) = 93 + 9.62 = 102.62$$

---

> **20.** A population is normally distributed with a variance of $50$. $98.17\%$ of the values of the population are less than $27$. The mean of the population is \_\_\_\_\_\_\_\_.

* **Answer:** **`12.22`**
* **Solution:**
  With $\sigma = \sqrt{50} \approx 7.0711$ and $P(X < 27) = 0.9817 \implies z = 2.09$:
  $$2.09 = \frac{27 - \mu}{\sqrt{50}} \implies \mu = 27 - 2.09\sqrt{50} \approx 12.22$$

---

> **21.** A population is normally distributed with a mean of $340$ and a standard deviation of $55$. $10.93\%$ of values in the population are less than \_\_\_\_\_\_\_\_.

* **Answer:** **`272.35`** (`272.34` exact in R)
* **Solution:**
  $P(X < x) = 0.1093 \implies z = -1.23$:
  $$x = \mu + z\sigma = 340 + (-1.23)(55) = 340 - 67.65 = 272.35$$

## Normal Approximation to the Binomial — Questions 22–33

> **22.** In working a binomial distribution problem by using the normal distribution, the interval, \_\_\_\_\_\_\_\_, should lie between $0$ and $n$.

* **Answer:** **`μ ± 3σ`**
* **Solution:**
  Ensuring $\mu \pm 3\sigma$ (roughly $99.7\%$ of the normal curve) falls inside $[0, n]$ prevents impossible negative counts or counts exceeding $n$ (equivalently satisfied by $np \ge 5$ and $nq \ge 5$).

---

> **23.** A binomial distribution problem has an $n$ of $10$ and a $p$ of $.20$. This problem \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ be worked by the normal distribution because of the size of $n$ and $p$.

* **Answer:** **`Cannot`**
* **Solution:**
  Here $np = 10(0.20) = 2 < 5$, violating the $np \ge 5$ condition.

---

> **24.** A binomial distribution problem has an $n$ of $15$ and a $p$ of $.60$. This problem \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ be worked by the normal distribution because of the size of $n$ and $p$.

* **Answer:** **`Can`**
* **Solution:**
  Both $np = 15(0.60) = 9 \ge 5$ and $nq = 15(0.40) = 6 \ge 5$ are satisfied.

---

> **25.** A binomial distribution problem has an $n$ of $30$ and a $p$ of $.35$. A researcher wants to determine the probability of $x$ being greater than $13$ and to use the normal distribution to work the problem. After correcting for continuity, the value of $x$ that he/she will be solving for is \_\_\_\_\_\_\_\_.

* **Answer:** **`13.5`**
* **Solution:**
  For discrete $X > 13$ (i.e., $X \ge 14$), exclude the bar at $13$ by starting at $13.5$: $P(Y \ge 13.5)$.

---

> **26.** A binomial distribution problem has an $n$ of $48$ and a $p$ of $.80$. A researcher wants to determine the probability of $x$ being less than or equal to $35$ and wants to work the problem using the normal distribution. After correcting for continuity, the value of $x$ that he/she will be solving for is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`35.5`**
* **Solution:**
  To include the bar centered at $35$ in $X \le 35$, extend the upper boundary to $35.5$: $P(Y \le 35.5)$.

---

> **27.** A binomial distribution problem has an $n$ of $60$ and a $p$ value of $.72$. A researcher wants to determine the probability of $x$ being exactly $45$ and use the normal distribution to work the problem. After correcting for continuity, he/she will be solving for the area between \_\_\_\_\_\_\_\_ and \_\_\_\_\_\_\_\_.

* **Answer:** **`44.5` and `45.5`**
* **Solution:**
  The integer bar at $45$ spans the continuous interval $[44.5, 45.5]$.

---

> **28.** A binomial distribution problem has an $n$ of $27$ and a $p$ of $.53$. If this problem were converted to a normal distribution problem, the mean of the distribution would be \_\_\_\_\_\_\_\_. The standard deviation of the distribution would be \_\_\_\_\_\_\_\_.

* **Answer:** **`14.31`, `2.59`** (`2.5934` exact)
* **Solution:**
  $$\mu = np = 27(0.53) = 14.31, \qquad \sigma = \sqrt{npq} = \sqrt{27(0.53)(0.47)} = \sqrt{6.7257} \approx 2.5934$$

---

> **29.** A binomial distribution problem has an $n$ of $113$ and a $p$ of $.29$. If this problem were converted to a normal distribution problem, the mean of the distribution would be \_\_\_\_\_\_\_\_. The standard deviation of the distribution would be \_\_\_\_\_\_\_\_.

* **Answer:** **`32.77`, `4.82`** (`4.8236` exact)
* **Solution:**
  $$\mu = 113(0.29) = 32.77, \qquad \sigma = \sqrt{113(0.29)(0.71)} = \sqrt{23.2667} \approx 4.8236$$

---

> **30.** A binomial distribution problem is to determine the probability that $x$ is less than $22$ when the sample size is $40$ and the value of $p$ is $.50$. Using the normal distribution to work this problem produces a probability of \_\_\_\_\_\_\_\_.

* **Answer:** **`0.6808`** *(Z-table rounded $z = 0.47$)*; **`0.6824`** *(exact in R)*
* **Solution:**
  Here $\mu = 40(0.50) = 20$ and $\sigma = \sqrt{40(0.50)(0.50)} = \sqrt{10} \approx 3.1623$. With continuity correction, $P(X < 22) \approx P(Y \le 21.5)$:
  $$z = \frac{21.5 - 20}{3.1623} \approx 0.4743 \approx 0.47 \implies P(Z \le 0.47) = 0.6808$$
  In R without intermediate rounding: `pnorm(21.5, 20, sqrt(10)) = 0.6824`.

---

> **31.** A binomial distribution problem is to determine the probability that $x$ is exactly $14$ when the sample size is $20$ and the value of $p$ is $.60$. Using the normal distribution to work this problem produces a probability of \_\_\_\_\_\_\_\_.

* **Answer:** **`0.1212`** *(Z-table rounded $z$)*; **`0.1199`** *(exact in R)*
* **Solution:**
  With $\mu = 20(0.60) = 12$ and $\sigma = \sqrt{20(0.60)(0.40)} = \sqrt{4.8} \approx 2.1909$:
  $$z_1 = \frac{13.5 - 12}{2.1909} \approx 0.68, \qquad z_2 = \frac{14.5 - 12}{2.1909} \approx 1.14$$
  Z-table: $P(0.68 \le Z \le 1.14) = 0.8729 - 0.7517 = 0.1212$. Exact in R: `0.1199`.

---

> **32.** A binomial distribution problem is to determine the probability that $x$ is greater than or equal to $18$ when the sample size is $30$ and the value of $p$ is $.55$. Using the normal distribution to work this problem produces a probability of \_\_\_\_\_\_\_\_.

* **Answer:** **`0.3557`** *(Z-table rounded $z = 0.37$)*; **`0.3568`** *(exact in R)*
* **Solution:**
  With $\mu = 30(0.55) = 16.5$ and $\sigma = \sqrt{30(0.55)(0.45)} = \sqrt{7.425} \approx 2.7249$:
  $$P(X \ge 18) \approx P(Y \ge 17.5), \qquad z = \frac{17.5 - 16.5}{2.7249} \approx 0.3670 \approx 0.37$$
  Z-table: $P(Z \ge 0.37) = 0.3557$. Exact in R: `0.3568`.

---

> **33.** A binomial distribution problem is to determine the probability that $x$ is greater than $10$ when the sample size is $20$ and the value of $p$ is $.60$. Using the normal distribution to work this problem produces a probability of \_\_\_\_\_\_\_\_. If this problem had been worked using the binomial tables, the obtained probability would have been \_\_\_\_\_\_\_\_. The difference in answers using these two techniques is \_\_\_\_\_\_\_\_.

* **Answer:** **`0.7517`, `0.7550`, `0.0033`** *(table rounded)*; **`0.7532`, `0.7553`, `0.0021`** *(exact in R)*
* **Solution:**
  With $\mu = 12$ and $\sigma = \sqrt{4.8} \approx 2.1909$, $P(X > 10) = P(X \ge 11) \approx P(Y \ge 10.5)$:

  Exact binomial $P(X > 10) = 1 - P(X \le 10) = 0.7553$ (4-decimal table: $0.7550$). Difference: $0.7550 - 0.7517 = 0.0033$ (exact R difference: $0.0021$).

```r
print(sprintf("Q28 -> mu = %.2f, sd = %.4f | Q29 -> mu = %.2f, sd = %.4f",
              27 * 0.53, sqrt(27 * 0.53 * 0.47), 113 * 0.29, sqrt(113 * 0.29 * 0.71)))
print(sprintf("Q30 normal approx: %.4f | Q31 normal approx: %.4f | Q32 normal approx: %.4f",
              pnorm(21.5, 40 * 0.5, sqrt(40 * 0.5 * 0.5)),
              pnorm(14.5, 20 * 0.6, sqrt(20 * 0.6 * 0.4)) - pnorm(13.5, 20 * 0.6, sqrt(20 * 0.6 * 0.4)),
              pnorm(17.5, 30 * 0.55, sqrt(30 * 0.55 * 0.45), lower.tail = FALSE)))
norm_33 <- pnorm(10.5, 20 * 0.6, sqrt(20 * 0.6 * 0.4), lower.tail = FALSE)
binom_33 <- pbinom(10, 20, 0.6, lower.tail = FALSE)
print(sprintf("Q33 -> Normal: %.4f | Binomial: %.4f | Diff: %.4f", norm_33, binom_33, abs(binom_33 - norm_33)))
```

## The Exponential Distribution — Questions 34–44

> **34.** The exponential distribution is a \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ distribution.

* **Answer:** **`Continuous`**
* **Solution:**
  Unlike the Poisson distribution (which counts discrete occurrences), the exponential distribution measures continuous waiting times or spaces between events.

---

> **35.** The exponential distribution is closely related to the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ distribution.

* **Answer:** **`Poisson`**
* **Solution:**
  When the count of arrivals in an interval follows a Poisson process with rate $\lambda$, the interarrival time between consecutive arrivals follows an Exponential distribution with rate $\lambda$.

---

> **36.** The exponential distribution is skewed to the \_\_\_\_\_\_\_\_.

* **Answer:** **`Right`**
* **Solution:**
  Because density starts at $\lambda$ at $x = 0$ and decays monotonically over $[0, \infty)$, the exponential distribution is strictly positively (right) skewed.

---

> **37.** Suppose random arrivals occur at a rate of $5$ per minute. Assuming that random arrivals are Poisson distributed, the probability of there being at least $30$ seconds between arrivals is \_\_\_\_\_\_\_\_.

* **Answer:** **`0.0821`**
* **Solution:**
  Convert $30$ seconds to $x_0 = 0.5$ minutes to match $\lambda = 5$/minute:
  $$P(X \ge 0.5) = e^{-5(0.5)} = e^{-2.5} \approx 0.0821$$

---

> **38.** Suppose random arrivals occur at a rate of $1$ per hour. Assuming that random arrivals are Poisson distributed, the probability of there being less than $2$ hours between arrivals is \_\_\_\_\_\_\_\_.

* **Answer:** **`0.8647`**
* **Solution:**
  With $\lambda = 1$/hour and $x_0 = 2$ hours:
  $$P(X < 2) = 1 - e^{-1(2)} = 1 - e^{-2} \approx 0.8647$$

---

> **39.** Suppose random arrivals occur at a rate of $1.6$ every five minutes. Assuming that random arrivals are Poisson distributed, the probability of there being between three minutes and six minutes between arrivals is \_\_\_\_\_\_\_\_.

* **Answer:** **`0.2363`**
* **Solution:**
  Convert $\lambda$ to a per-minute rate: $\lambda = \frac{1.6}{5} = 0.32$ arrivals/minute:
  $$P(3 \le X \le 6) = \left(1 - e^{-0.32(6)}\right) - \left(1 - e^{-0.32(3)}\right) = e^{-0.96} - e^{-1.92} \approx 0.3829 - 0.1466 = 0.2363$$

---

> **40.** Suppose that the mean time between arrivals is $40$ seconds and that random arrivals are Poisson distributed. The probability that at least one minute passes between two arrivals is \_\_\_\_\_\_\_\_. The probability that at least two minutes pass between two arrivals is \_\_\_\_\_\_\_\_.

* **Answer:** **`0.2231`, `0.0498`**
* **Solution:**
  Mean interarrival time $\mu = 40$ seconds $= 2/3$ minute $\implies \lambda = 1.5$ arrivals/minute (or $\lambda = 1/40$ per second):
  * $P(X \ge 60\text{ sec}) = e^{-(1/40)(60)} = e^{-1.5} \approx 0.2231$
  * $P(X \ge 120\text{ sec}) = e^{-(1/40)(120)} = e^{-3.0} \approx 0.0498$

---

> **41.** Suppose that the mean time between arrivals is ten minutes and that random arrivals are Poisson distributed. The probability that no more than seven minutes pass between two arrivals is \_\_\_\_\_\_\_\_.

* **Answer:** **`0.5034`**
* **Solution:**
  With $\mu = 10$ minutes, $\lambda = 1/10 = 0.10$/minute:
  $$P(X \le 7) = 1 - e^{-0.10(7)} = 1 - e^{-0.7} \approx 0.5034$$

---

> **42.** The mean of an exponential distribution equals \_\_\_\_\_\_\_\_.

* **Answer:** **`1/λ`**
* **Solution:**
  Both the mean and the standard deviation of an exponential distribution equal the reciprocal of the rate parameter: $\mu = \sigma = \frac{1}{\lambda}$.

---

> **43.** Suppose that random arrivals are Poisson distributed with an average arrival of $2.4$ per five minutes. The associated exponential distribution would have a mean of \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ and a standard deviation of \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`2.08 Minutes`, `2.08 Minutes`** (`2.0833` min exact)
* **Solution:**
  Rate per minute is $\lambda = \frac{2.4}{5} = 0.48$/minute, so:
  $$\mu = \sigma = \frac{1}{0.48} = \frac{5}{2.4} \approx 2.0833 \text{ minutes}$$

---

> **44.** An exponential distribution has an average interarrival time of $25$ minutes. The standard deviation of this distribution is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`25 Minutes`**
* **Solution:**
  Because $\mu = \sigma = 1/\lambda$ for every exponential distribution, the standard deviation equals the mean ($25$ minutes).

```r
print(sprintf("Q37: %.4f | Q38: %.4f | Q39: %.4f",
              pexp(0.5, rate = 5, lower.tail = FALSE),
              pexp(2, rate = 1, lower.tail = TRUE),
              pexp(6, rate = 1.6 / 5) - pexp(3, rate = 1.6 / 5)))
print(sprintf("Q40a: %.4f | Q40b: %.4f | Q41: %.4f | Q43 mean=sd: %.4f",
              pexp(1, rate = 60 / 40, lower.tail = FALSE),
              pexp(2, rate = 60 / 40, lower.tail = FALSE),
              pexp(7, rate = 1 / 10, lower.tail = TRUE),
              5 / 2.4))
```

> **4.** The directory or map from which a sample is taken is called the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

## Central Limit Theorem & Sampling Distributions — Questions 21–28

> **21.** The central limit theorem states that if $n$ is large enough, the sample means are \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ distributed regardless of the shape of the population.

* **Answer:** **`Normally`**
* **Solution:**
  Even when the underlying population is skewed, uniform, or multimodal, the sampling distribution of $\bar{x}$ approaches normality as $n$ grows (typically $n \ge 30$).

---

> **22.** According to the central limit theorem, the mean of the sample means for a given size of sample is equal to the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ \_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`Population Mean`** ($\mu_{\bar{x}} = \mu$)
* **Solution:**
  The sample mean $\bar{x}$ is an unbiased estimator of $\mu$: $E(\bar{x}) = \mu$.

---

> **23.** According to the central limit theorem, the standard deviation of sample means for a given size of sample equals \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`σ / √n`**
* **Solution:**
  The standard error of the mean is $\sigma_{\bar{x}} = \frac{\sigma}{\sqrt{n}}$.

---

> **24.** If samples are being drawn from a known population size, the $z$ formula for sample means includes a \_\_\_\_\_\_\_\_\_\_ \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ factor.

* **Answer:** **`Finite Correction`** (finite population correction factor, $\sqrt{\frac{N - n}{N - 1}}$)
* **Solution:**
  When sampling without replacement from a finite population where $n > 0.05N$, multiply $\frac{\sigma}{\sqrt{n}}$ by $\sqrt{\frac{N - n}{N - 1}}$.

---

> **25.** Suppose a population has a mean of $90$ and a standard deviation of $27$. If a random sample of size $49$ is drawn from the population, the probability of drawing a sample with a mean of more than $95$ is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`0.0968`** *(Z-table rounded $z = 1.30$)*; **`0.0974`** *(exact in R)*
* **Solution:**
  Standard error is $\sigma_{\bar{x}} = \frac{27}{\sqrt{49}} = \frac{27}{7} \approx 3.8571$:
  $$z = \frac{95 - 90}{27 / 7} = \frac{35}{27} \approx 1.2963 \approx 1.30 \implies P(Z > 1.30) = 0.0968$$
  Exact in R: `pnorm(95, 90, 27 / sqrt(49), lower.tail = FALSE) = 0.0974`.

---

> **26.** Suppose a population has a mean of $455$ and a variance of $900$. If a random sample of size $65$ is drawn from the population, the probability that the sample mean is between $448$ and $453$ is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`0.2645`** *(Z-table rounded $z$)*; **`0.2655`** *(exact in R)*
* **Solution:**
  With $\sigma = \sqrt{900} = 30$ and $\sigma_{\bar{x}} = \frac{30}{\sqrt{65}} \approx 3.7210$:
  $$z_1 = \frac{448 - 455}{3.7210} \approx -1.88, \qquad z_2 = \frac{453 - 455}{3.7210} \approx -0.54$$
  Z-table: $P(-1.88 \le Z \le -0.54) = 0.2946 - 0.0301 = 0.2645$. Exact in R: `0.2655`.

---

> **27.** Suppose $.60$ of the population possess a given characteristic. If a random sample of size $300$ is drawn from the population, then the probability that $.53$ or fewer of the sample possess the characteristic is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`0.0068`** *(Z-table rounded $z = -2.47$)*; **`0.0067`** *(exact in R)*
* **Solution:**
  Here $np = 180 \ge 5$, $nq = 120 \ge 5$, and $\sigma_{\hat{p}} = \sqrt{\frac{0.60(0.40)}{300}} = \sqrt{0.0008} \approx 0.028284$:
  $$z = \frac{0.53 - 0.60}{0.028284} \approx -2.4749 \approx -2.47 \implies P(Z \le -2.47) = 0.0068$$
  Exact in R: `pnorm(0.53, 0.60, sqrt(0.60 * 0.40 / 300)) = 0.006664`.

---

> **28.** Suppose $.36$ of a population possess a given characteristic. If a random sample of size $1200$ is drawn from the population, then the probability that less than $480$ possess that characteristic in the sample is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`0.9981`**
* **Solution:**
  The threshold sample proportion is $\hat{p} = \frac{480}{1200} = 0.40$, with $\sigma_{\hat{p}} = \sqrt{\frac{0.36(0.64)}{1200}} = \sqrt{0.000192} \approx 0.013856$:
  $$z = \frac{0.40 - 0.36}{0.013856} \approx 2.8868 \approx 2.89 \implies P(Z < 2.89) = 0.9981$$

```r
print(sprintf("Q25: %.4f | Q26: %.4f",
              pnorm(95, 90, 27 / sqrt(49), lower.tail = FALSE),
              pnorm(453, 455, sqrt(900) / sqrt(65)) - pnorm(448, 455, sqrt(900) / sqrt(65))))
print(sprintf("Q27: %.4f | Q28: %.4f",
              pnorm(0.53, 0.60, sqrt(0.60 * 0.40 / 300)),
              pnorm(480 / 1200, 0.36, sqrt(0.36 * 0.64 / 1200))))
```

## Exercise 3: Random Sampling Variability & Sample Size
Generate a sample of size $50$ from a normal distribution with a mean of $100$ and a standard deviation of $4$. Compute the sample mean, sample standard deviation, and standard error of the mean. Generate a second sample of size $50$ from the same normal population and compare. Are the sample means and sample standard deviations identical? Why or why not? Now repeat this process generating a sample of size $5000$ and compare to the previous samples.

```r
set.seed(1234)
my_first_sample <- rnorm(n = 50, mean = 100, sd = 4)
std_error1 <- 4 / sqrt(50)

my_second_sample <- rnorm(n = 50, mean = 100, sd = 4)
std_error2 <- 4 / sqrt(50)

my_third_sample <- rnorm(n = 5000, mean = 100, sd = 4)
std_error3 <- 4 / sqrt(5000)

print(sprintf("Sample 1 (n = 50)   -> Mean: %.5f | SD: %.6f | SE: %.7f",
              mean(my_first_sample), sd(my_first_sample), std_error1))
print(sprintf("Sample 2 (n = 50)   -> Mean: %.5f | SD: %.6f | SE: %.7f",
              mean(my_second_sample), sd(my_second_sample), std_error2))
print(sprintf("Sample 3 (n = 5000) -> Mean: %.5f | SD: %.6f | SE: %.8f",
              mean(my_third_sample), sd(my_third_sample), std_error3))
```

* Increasing the sample size $100$-fold from $n = 50$ to $n = 5000$ reduces the standard error by a factor of $\sqrt{100} = 10$ (from $0.5657$ to $0.0566$), causing the sample mean ($99.9920$) and sample standard deviation ($3.9669$) to converge tightly toward the true population parameters ($\mu = 100$, $\sigma = 4$).

## Exercise 4: Normal Approximation to the Binomial (Biased Coin)
Assume a biased coin when flipped will generate heads one third of the time ($p = 1/3$). Estimate the probability of getting **at least $250$ heads** out of $n = 600$ flips using the normal distribution approximation (with continuity correction). Compare to the exact probability using the binomial distribution.

```r
n_coin <- 600
p_coin <- 1 / 3
x_target <- 250

exact_at_least_250 <- pbinom(q = x_target - 1, size = n_coin, prob = p_coin, lower.tail = FALSE)
exact_gt_250 <- pbinom(q = x_target, size = n_coin, prob = p_coin, lower.tail = FALSE)

x_correct <- x_target - 0.5
z_coin <- (x_correct - n_coin * p_coin) / sqrt(n_coin * p_coin * (1 - p_coin))
approx_at_least_250 <- pnorm(z_coin, mean = 0, sd = 1, lower.tail = FALSE)

print(sprintf("Exact Binomial P(X >= 250) [pbinom(249, lower.tail = FALSE)]: %.6f", exact_at_least_250))
print(sprintf("Exact Binomial P(X > 250)  [pbinom(250, lower.tail = FALSE)]: %.6f", exact_gt_250))
print(sprintf("Normal Approx  P(X >= 250) [z at 249.5 = %.4f]: %.6f", z_coin, approx_at_least_250))
```

**Interpretation (Exercise 4):** Because $np = 200 \ge 5$ and $nq = 400 \ge 5$, the normal approximation with continuity correction ($249.5$, yielding $z = 4.29$, $P \approx 0.000009$) closely matches the exact binomial probability ($P(X \ge 250) = 0.000013$). Both confirm that observing at least $250$ heads in $600$ flips of a $1/3$-biased coin is extraordinarily rare.
