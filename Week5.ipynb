# Chapter 8 study questions (Estimation for single populations)

The following 34 study questions from `Chapter8PracticeProblems.pdf` cover point vs. interval estimation, confidence intervals for $\mu$ ($\sigma$ known and $\sigma$ unknown), confidence intervals for $p$, confidence intervals for $\sigma^2$, and minimum sample size determination for $\mu$ and $p$.

## Point vs. Interval Estimates & Critical $z$-Values — Questions 1–5

> **1.** When a statistic taken from the sample is used to estimate a population parameter, it is called a(n) \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ estimate.

* **Answer:** **`Point`**
* **Solution:**
  A single sample statistic (such as $\bar{x}$, $\hat{p}$, or $s^2$) used as a best single-number guess for an unknown population parameter is a **point estimate**.

---

> **2.** When a range of values is used to estimate a population parameter, it is called a(n) \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ estimate.

* **Answer:** **`Interval`**
* **Solution:**
  A range constructed around a point estimate (a confidence interval) within which the population parameter is expected to lie at a specified confidence level is an **interval estimate**.

---

> **3.** The $z$ value associated with a two-sided $90\%$ confidence interval is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`1.645`**
* **Solution:**
  For a $90\%$ confidence interval, $\alpha = 0.10$ and $\alpha/2 = 0.05$, so $z_{0.05} = \Phi^{-1}(0.95) \approx 1.645$.

---

> **4.** The $z$ value associated with a two-sided $95\%$ confidence interval is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`1.96`**
* **Solution:**
  For a $95\%$ confidence interval, $\alpha = 0.05$ and $\alpha/2 = 0.025$, so $z_{0.025} = \Phi^{-1}(0.975) \approx 1.960$.

---

> **5.** The $z$ value associated with a two-sided $80\%$ confidence interval is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`1.28`**
* **Solution:**
  For an $80\%$ confidence interval, $\alpha = 0.20$ and $\alpha/2 = 0.10$, so $z_{0.10} = \Phi^{-1}(0.90) \approx 1.282$ (rounded to `1.28` in $z$-tables).

```r
print(sprintf("Q3 (90%% CI z) = %.4f", qnorm(0.95)))
print(sprintf("Q4 (95%% CI z) = %.4f", qnorm(0.975)))
print(sprintf("Q5 (80%% CI z) = %.4f", qnorm(0.90)))
```

## Confidence Intervals for $\mu$ ($\sigma$ Known) — Questions 6–9

> **6.** Suppose a random sample of $40$ is selected from a population with a standard deviation of $13$. If the sample mean is $118$, the $98\%$ confidence interval to estimate the population mean is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`113.2 < μ < 122.8`**
* **Solution:**
  With $n = 40$, $\sigma = 13$, $\bar{x} = 118$, and $z_{0.01} = 2.3263$:
  $$118 \pm 2.3263 \left(\frac{13}{\sqrt{40}}\right) = 118 \pm 4.7817 \implies (113.2, \; 122.8)$$

---

> **7.** Suppose a random sample of size $75$ is selected from a population with a standard deviation of $6.4$. The sample yields a mean of $26$. From this information, the $90\%$ confidence interval to estimate the population mean can be computed as \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`24.8 < μ < 27.2`**
* **Solution:**
  With $n = 75$, $\sigma = 6.4$, $\bar{x} = 26$, and $z_{0.05} = 1.6449$:
  $$26 \pm 1.6449 \left(\frac{6.4}{\sqrt{75}}\right) = 26 \pm 1.2156 \implies (24.8, \; 27.2)$$

---

> **8.** The following random sample of numbers are drawn from a population:
> $45, 61, 55, 43, 49, 60, 62, 53, 57, 44, 39, 48, 57, 40, 61, 62, 45, 39, 38, 56, 55, 59, 63, 50, 41, 39, 45, 47, 56, 51, 61, 39, 36, 57$.
> Assume that the population standard deviation is $8.62$. From these data, a $99\%$ confidence interval to estimate the population mean can be computed as \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`46.6 < μ < 54.2`**
* **Solution:**
  There are $n = 34$ observations with $\bar{x} = 50.3824$, $\sigma = 8.62$, and $z_{0.005} = 2.5758$:
  $$50.3824 \pm 2.5758 \left(\frac{8.62}{\sqrt{34}}\right) = 50.3824 \pm 3.8079 \implies (46.6, \; 54.2)$$

---

> **9.** A random sample of $63$ items is selected from a population of $400$ items. The sample mean is $211$. The population standard deviation is $48$. From this information, a $95\%$ confidence interval to estimate the population mean can be computed as \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`200.1 < μ < 221.9`**
* **Solution:**
  Because $n / N = 63 / 400 = 0.1575 > 0.05$, we apply the **Finite Population Correction (FPC)** factor $\sqrt{\frac{N - n}{N - 1}} = \sqrt{\frac{400 - 63}{399}} = 0.9190$:
  $$211 \pm 1.9600 \left(\frac{48}{\sqrt{63}}\right)\sqrt{\frac{337}{399}} = 211 \pm 10.8927 \implies (200.1, \; 221.9)$$

```r
q6_me <- qnorm(0.99) * 13 / sqrt(40)
print(sprintf("Q6 -> 98%% CI: [%.1f, %.1f]", 118 - q6_me, 118 + q6_me))

q7_me <- qnorm(0.95) * 6.4 / sqrt(75)
print(sprintf("Q7 -> 90%% CI: [%.1f, %.1f]", 26 - q7_me, 26 + q7_me))

q8_data <- c(45, 61, 55, 43, 49, 60, 62, 53, 57, 44, 39, 48, 57, 40, 61, 62, 45, 39,
             38, 56, 55, 59, 63, 50, 41, 39, 45, 47, 56, 51, 61, 39, 36, 57)
q8_mean <- mean(q8_data)
q8_me <- qnorm(0.995) * 8.62 / sqrt(length(q8_data))
print(sprintf("Q8 -> n = %d | mean = %.4f | 99%% CI: [%.1f, %.1f]",
              length(q8_data), q8_mean, q8_mean - q8_me, q8_mean + q8_me))

fpc_9 <- sqrt((400 - 63) / (400 - 1))
q9_me <- qnorm(0.975) * (48 / sqrt(63)) * fpc_9
print(sprintf("Q9 -> FPC = %.4f | 95%% CI: [%.1f, %.1f]",
              fpc_9, 211 - q9_me, 211 + q9_me))
```

## Confidence Intervals for $\mu$ ($\sigma$ Unknown, $t$-Distribution) — Questions 10–17

> **10.** Generally, when estimating a population mean and the population standard deviation is not known, you should use the \_\_\_\_\_ statistic.

* **Answer:** **`t`**
* **Solution:**
  When $\sigma$ is unknown and estimated by the sample standard deviation $s$, the standardized mean follows **Student's $t$-distribution** with $df = n - 1$.

---

> **11.** The $t$ test was developed by \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`William S. Gosset`**
* **Solution:**
  William Sealy Gosset published the $t$-distribution in 1908 under the pseudonym **"Student"** while working as a chemist and statistician at the Guinness Brewery in Dublin.

---

> **12.** In order to find values in the $t$ distribution table, you must convert the sample size or sizes to \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`Degrees of Freedom`**
* **Solution:**
  For a single-sample $t$-interval, degrees of freedom are $df = n - 1$.

---

> **13.** The table $t$ value associated with $10$ degrees of freedom and used to compute a $95\%$ confidence interval is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`2.228`**
* **Solution:**
  For $df = 10$ and $\alpha/2 = 0.025$, $t_{0.025,\, 10} = \text{qt}(0.975, 10) = 2.2281$.

---

> **14.** The table $t$ value associated with $18$ degrees of freedom and used to compute a $99\%$ confidence interval is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`2.878`**
* **Solution:**
  For $df = 18$ and $\alpha/2 = 0.005$, $t_{0.005,\, 18} = \text{qt}(0.995, 18) = 2.8784$.

---

> **15.** A researcher is interested in estimating the mean value for a population. She takes a random sample of $17$ items and computes a sample mean of $224$ and a sample standard deviation of $32$. She decides to construct a $98\%$ confidence interval to estimate the mean. The degrees of freedom associated with this problem are \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_. It can be assumed that these values are normally distributed in the population.

* **Answer:** **`16`**
* **Solution:**
  $df = n - 1 = 17 - 1 = 16$.

---

> **16.** The table $t$ value used to construct the confidence interval in question 15 is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`2.583`**
* **Solution:**
  For a $98\%$ confidence interval ($\alpha/2 = 0.01$) with $df = 16$, $t_{0.01,\, 16} = \text{qt}(0.99, 16) = 2.5835$.

---

> **17.** The confidence interval resulting from the data in question 15 is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`203.95 < μ < 244.05`**
* **Solution:**
  $$224 \pm 2.5835 \left(\frac{32}{\sqrt{17}}\right) = 224 \pm 20.0508 \implies (203.95, \; 244.05)$$

```r
print(sprintf("Q13 -> t(0.975, df=10) = %.3f", qt(0.975, df = 10)))
print(sprintf("Q14 -> t(0.995, df=18) = %.3f", qt(0.995, df = 18)))
df15 <- 17 - 1
t16 <- qt(0.99, df = df15)
me17 <- t16 * 32 / sqrt(17)
print(sprintf("Q15-Q17 -> df = %d | t = %.3f | 98%% CI: [%.2f, %.2f]",
              df15, t16, 224 - me17, 224 + me17))
```

## Confidence Intervals for a Population Proportion ($p$) — Questions 18–21

> **18.** A researcher wants to estimate the proportion of the population which possesses a given characteristic. A random sample of size $800$ is taken resulting in $380$ items which possess the characteristic. The point estimate for this population proportion is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`0.475`**
* **Solution:**
  $\hat{p} = \frac{x}{n} = \frac{380}{800} = 0.475$.

---

> **19.** A researcher wants to estimate the proportion of a population which possesses a given characteristic. A random sample of size $1250$ is taken and $.67$ of the sample possess the characteristic. The $90\%$ confidence interval to estimate the population proportion is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`0.648 < p < 0.692`**
* **Solution:**
  With $n = 1250$, $\hat{p} = 0.67$, $\hat{q} = 0.33$, and $z_{0.05} = 1.6449$:
  $$0.67 \pm 1.6449 \sqrt{\frac{0.67(0.33)}{1250}} = 0.67 \pm 0.0219 \implies (0.648, \; 0.692)$$

---

> **20.** A random sample of $255$ items from a population results in $44\%$ possessing a given characteristic. Using this information, the researcher constructs a $99\%$ confidence interval to estimate the population proportion. The resulting confidence interval is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`0.36 < p < 0.52`**
* **Solution:**
  With $n = 255$, $\hat{p} = 0.44$, $\hat{q} = 0.56$, and $z_{0.005} = 2.5758$:
  $$0.44 \pm 2.5758 \sqrt{\frac{0.44(0.56)}{255}} = 0.44 \pm 0.0801 \implies (0.36, \; 0.52)$$

---

> **21.** What proportion of a population possesses a given characteristic? To estimate this, a random sample of $1700$ people are interviewed from the population. Seven hundred and fourteen of the people sampled possess the characteristic. Using this information, the researcher computes an $88\%$ confidence interval to estimate the proportion of the population who possess the given characteristic. The resulting confidence interval is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`0.401 < p < 0.439`**
* **Solution:**
  Here $\hat{p} = \frac{714}{1700} = 0.42$, $\hat{q} = 0.58$, and for an $88\%$ CI ($\alpha = 0.12$, $\alpha/2 = 0.06$), $z_{0.06} = 1.5548$:
  $$0.42 \pm 1.5548 \sqrt{\frac{0.42(0.58)}{1700}} = 0.42 \pm 0.0186 \implies (0.401, \; 0.439)$$

```r
print(sprintf("Q18 -> p_hat = %.3f", 380 / 800))

me19 <- qnorm(0.95) * sqrt(0.67 * 0.33 / 1250)
print(sprintf("Q19 -> 90%% CI: [%.3f, %.3f]", 0.67 - me19, 0.67 + me19))

me20 <- qnorm(0.995) * sqrt(0.44 * 0.56 / 255)
print(sprintf("Q20 -> 99%% CI: [%.2f, %.2f] (exact: [%.4f, %.4f])",
              0.44 - me20, 0.44 + me20, 0.44 - me20, 0.44 + me20))

p21 <- 714 / 1700
me21 <- qnorm(0.94) * sqrt(p21 * (1 - p21) / 1700)
print(sprintf("Q21 -> p_hat = %.2f | 88%% CI: [%.3f, %.3f]", p21, p21 - me21, p21 + me21))
```

## Confidence Intervals for a Population Variance ($\sigma^2$) — Questions 22–25

> **22.** A confidence interval to estimate the population variance can be constructed by using the sample variance and the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ distribution.

* **Answer:** **`Chi-square`**
* **Solution:**
  Assuming a normal population, $\frac{(n - 1)s^2}{\sigma^2}$ follows a **Chi-square ($\chi^2$)** distribution with $df = n - 1$.

---

> **23.** Suppose we want to construct a confidence interval to estimate a population variance. A sample variance is computed from a sample of $14$ items. To construct a $95\%$ confidence interval, the chi-square table values are \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ and \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`5.00874` and `24.7356`**
* **Solution:**
  With $n = 14$ ($df = 13$) and $\alpha = 0.05$:
  * Lower-tail critical value: $\chi^2_{0.975,\, 13} = \text{qchisq}(0.025, 13) = 5.00875$ (table key: `5.00874`)
  * Upper-tail critical value: $\chi^2_{0.025,\, 13} = \text{qchisq}(0.975, 13) = 24.7356$

---

> **24.** We want to estimate a population variance. A sample of $9$ items produces a sample standard deviation of $4.29$. The point estimate of the population variance is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`s² = 18.4041`**
* **Solution:**
  $s^2 = (4.29)^2 = 18.4041$.

---

> **25.** In an effort to estimate the population variance, a sample of $12$ items is taken. The sample variance is $21.96$. Using this information, it can be determined that the $90\%$ confidence interval is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`12.277 < σ² < 52.802`**
* **Solution:**
  With $n = 12$ ($df = 11$), $s^2 = 21.96$, $\chi^2_{0.95,\, 11} = 4.5748$, and $\chi^2_{0.05,\, 11} = 19.6751$:
  $$\frac{11(21.96)}{19.6751} \le \sigma^2 \le \frac{11(21.96)}{4.5748} \implies 12.277 \le \sigma^2 \le 52.802$$

```r
print(sprintf("Q23 -> Lower chi^2: %.5f | Upper chi^2: %.4f",
              qchisq(0.025, df = 13), qchisq(0.975, df = 13)))
print(sprintf("Q24 -> Sample variance s^2 = %.4f", 4.29^2))
df25 <- 12 - 1
s2_25 <- 21.96
ci25 <- c(df25 * s2_25 / qchisq(0.95, df = df25),
          df25 * s2_25 / qchisq(0.05, df = df25))
print(sprintf("Q25 -> 90%% CI for sigma^2: [%.3f, %.3f]", ci25[1], ci25[2]))
```

## Determining Minimum Sample Size ($n$) — Questions 26–34

> **26.** In estimating the sample size necessary to estimate $\mu$, the error of estimation, $E$, is equal to \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`x̄ − μ`** *(more generally $|\bar{x} - \mu|$, the maximum allowable margin of error)*
* **Solution:**
  The estimation error is the difference between the sample mean and population mean, $E = \bar{x} - \mu$ (bounded in magnitude by $z_{\alpha/2}\frac{\sigma}{\sqrt{n}}$).

---

> **27.** In estimating sample size, if the population standard deviation is unknown, it can be estimated by using \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`1/4 Range`** *(or a pilot study)*
* **Solution:**
  By the Empirical Rule, approximately $95\%$ to $99.7\%$ of observations fall within $\mu \pm 2\sigma$ (width $4\sigma$), yielding the rule-of-thumb approximation $\sigma \approx \frac{1}{4}\text{Range}$.

---

> **28.** Suppose a researcher wants to conduct a study to estimate the population mean. He/she plans to use a $95\%$ level of confidence to estimate the mean and the population standard deviation is approximately $34$. The researcher wants the error to be no more than $4$. The sample size should be at least \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`278`**
* **Solution:**
  $$n = \left(\frac{1.9600 \times 34}{4}\right)^2 = 277.55 \implies 278$$

---

> **29.** A researcher wants to determine the sample size necessary to adequately conduct a study to estimate the population mean to within $5$ points. The range of population values is $80$ and the researcher plans to use a $90\%$ level of confidence. The sample size should be at least \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`44`**
* **Solution:**
  Approximating $\sigma \approx \frac{80}{4} = 20$ with $z_{0.05} = 1.6449$:
  $$n = \left(\frac{1.6449 \times 20}{5}\right)^2 = 43.29 \implies 44$$

---

> **30.** A study is going to be conducted in which a population mean will be estimated using a $92\%$ confidence interval. The estimate needs to be within $12$ of the actual population mean. The population variance is estimated to be around $2200$. The necessary sample size should be at least \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`47`**
* **Solution:**
  With $\sigma^2 = 2200$, $E = 12$, and $z_{0.04} = 1.7507$:
  $$n = \frac{(1.7507)^2 \times 2200}{12^2} = 46.82 \implies 47$$

---

> **31.** In estimating the sample size necessary to estimate $p$, if there is no good approximation for the value of $p$ available, the value of \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ should be used as an estimate of $p$ in the formula.

* **Answer:** **`0.50`**
* **Solution:**
  $p(1 - p)$ attains its maximum ($0.25$) at $p = 0.50$, guaranteeing a sufficiently large (conservative) sample size.

---

> **32.** A researcher wants to estimate the population proportion with a $95\%$ level of confidence. He/she estimates from previous studies that the population proportion is no more than $.30$. The researcher wants the estimate to have an error of no more than $.02$. The necessary sample size is at least \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`2,017`**
* **Solution:**
  Among values $p \le 0.30$, $p(1 - p)$ is maximized at $p = 0.30$:
  $$n = \frac{(1.9600)^2 (0.30)(0.70)}{0.02^2} = 2016.77 \implies 2,017$$

---

> **33.** A study will be conducted to estimate the population proportion. A level of confidence of $99\%$ will be used and an error of no more than $.05$ is desired. There is no knowledge as to what the population proportion will be. The size of sample should be at least \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`664`**
* **Solution:**
  Using conservative $p = 0.50$ and $z_{0.005} = 2.5758$:
  $$n = \frac{(2.5758)^2 (0.50)(0.50)}{0.05^2} = 663.49 \implies 664$$

---

> **34.** A researcher conducts a study to determine what the population proportion is for a given characteristic. It is believed from previous studies that the proportion of the population will be at least $.65$. The researcher wants to use a $98\%$ level of confidence. He/she also wants the error to be no more than $.03$. The sample size should be at least \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`1,369`** *(textbook PDF key prints **`1,373`** using table-rounded $z = 2.33$)*
* **Solution:**
  Among values $p \ge 0.65$, $p(1 - p)$ is maximized closest to $0.50$, i.e., at $p = 0.65$:
  * Exact R ($z_{0.01} = 2.32635$): $n = \frac{(2.32635)^2 (0.65)(0.35)}{0.03^2} = 1368.01 \implies 1,369$.
  * Textbook table ($z_{0.01} \approx 2.33$): $n = \frac{(2.33)^2 (0.65)(0.35)}{0.03^2} = 1372.32 \implies 1,373$.

```r
n28 <- (qnorm(0.975) * 34 / 4)^2
n29 <- (qnorm(0.95) * (80 / 4) / 5)^2
n30 <- qnorm(0.96)^2 * 2200 / 12^2
n32 <- qnorm(0.975)^2 * 0.30 * 0.70 / 0.02^2
n33 <- qnorm(0.995)^2 * 0.50 * 0.50 / 0.05^2
n34_exact <- qnorm(0.99)^2 * 0.65 * 0.35 / 0.03^2
n34_table <- 2.33^2 * 0.65 * 0.35 / 0.03^2

print(sprintf("Q28: %.2f -> %d | Q29: %.2f -> %d | Q30: %.2f -> %d",
              n28, ceiling(n28), n29, ceiling(n29), n30, ceiling(n30)))
print(sprintf("Q32: %.2f -> %d | Q33: %.2f -> %d",
              n32, ceiling(n32), n33, ceiling(n33)))
print(sprintf("Q34 exact: %.2f -> %d | Q34 (z=2.33): %.2f -> %d",
              n34_exact, ceiling(n34_exact), n34_table, ceiling(n34_table)))
```

## Exercise 1: $95\%$ Confidence Interval for $\mu$ ($\sigma$ Known)

Assume a random sample of size $100$ is drawn from a normal distribution with variance $1$. The average value of the sample is $50$. Find a $95\%$ confidence interval for the mean.

```r
n_ex1 <- 100
xbar_ex1 <- 50
sd_ex1 <- sqrt(1)
margin_ex1 <- qnorm(1 - (0.05 / 2)) * sd_ex1 / sqrt(n_ex1)
ci_ex1 <- c(xbar_ex1 - margin_ex1, xbar_ex1 + margin_ex1)
print(sprintf("95%% Confidence Interval for mu: [%.4f, %.4f]", ci_ex1[1], ci_ex1[2]))
```

**Interpretation (Exercise 1):** With $n = 100$, $\sigma = \sqrt{1} = 1$, and $\bar{x} = 50$, the margin of error is $1.96 \times \frac{1}{10} = 0.196$, producing a $95\%$ confidence interval of **$[49.8040, 50.1960]$**.

## Exercise 2: Required Sample Size for a Specified Interval Width

Assume the standard deviation for a normal distribution is equal to $100$ units. Also assume we want to estimate the unknown mean with a $95\%$ confidence interval of total width $8$ units. Calculate the sample size required.

```r
z_ex2 <- qnorm(0.025, mean = 0, sd = 1, lower.tail = FALSE)
E_ex2 <- 8 / 2
n_ex2_raw <- (z_ex2 * 100 / E_ex2)^2
print(sprintf("Exact n = %.4f | Rounded (round): %d | Rounded up (ceiling): %d",
              n_ex2_raw, round(n_ex2_raw), ceiling(n_ex2_raw)))
```

**Interpretation (Exercise 2):** A total confidence interval width of $8$ units corresponds to a margin of error $E = 8 / 2 = 4$ units. Using $z_{0.025} = 1.96$, $n = \left(\frac{1.96 \times 100}{4}\right)^2 = 2400.91$, so **$2{,}401$** observations are required.

## Exercise 3: $95\%$ Confidence Interval for Voter Approval Proportion

A random sample of $1600$ registered voters are contacted and asked a variety of questions. For one question, $60\%$ of the voters expressed approval and $40\%$ disapproval. Calculate a $95\%$ confidence interval for the proportion expressing approval.

```r
p_ex3 <- 0.60
n_ex3 <- 1600
z_ex3 <- qnorm(0.975)
wald_ex3 <- p_ex3 + c(-1, 1) * z_ex3 * sqrt(p_ex3 * (1 - p_ex3) / n_ex3)
prop_obj_cc <- prop.test(x = n_ex3 * p_ex3, n = n_ex3, alternative = "two.sided", conf.level = 0.95, correct = TRUE)
prop_obj_nocc <- prop.test(x = n_ex3 * p_ex3, n = n_ex3, alternative = "two.sided", conf.level = 0.95, correct = FALSE)

print(prop_obj_cc)
print(sprintf("Wald 95%% CI:            [%.6f, %.6f]", wald_ex3[1], wald_ex3[2]))
print(sprintf("prop.test (Yates CC):   [%.6f, %.6f]", prop_obj_cc$conf.int[1], prop_obj_cc$conf.int[2]))
print(sprintf("prop.test (Wilson):     [%.6f, %.6f]", prop_obj_nocc$conf.int[1], prop_obj_nocc$conf.int[2]))
```

**Interpretation (Exercise 3):** With $n = 1600$ and $x = 960$ approvals ($\hat{p} = 0.60$), the large-sample Wald interval ($[0.5760, 0.6240]$), `prop.test()` with Yates' continuity correction ($[0.5755, 0.6240]$), and Wilson Score interval ($[0.5758, 0.6237]$) all closely agree around $57.6\%$ to $62.4\%$.

## Exercise 4: Required Sample Size for Beverage Preference Proportion

A random sample of consumers are presented with two beverages in random order and asked which they prefer most. All the consumers expressed a preference. One beverage was preferred $85\%$ of the time. Use this number to determine how large a sample of consumers would be needed to generate a $95\%$ confidence interval with an overall width just less than $2\%$ (i.e. from $84\%$ to $86\%$)?

```r
p_ex4 <- 0.85
E_ex4 <- 0.02 / 2
z_ex4 <- qnorm(0.025, mean = 0, sd = 1, lower.tail = FALSE)
n_ex4_prior <- (z_ex4^2) * p_ex4 * (1 - p_ex4) / (E_ex4^2)
n_ex4_cons <- (z_ex4^2) * (0.25) / (E_ex4^2)

print(sprintf("Using p = 0.85 -> exact: %.4f | round: %d | ceiling: %d",
              n_ex4_prior, round(n_ex4_prior), ceiling(n_ex4_prior)))
print(sprintf("Conservative (p = 0.50) -> exact: %.4f | round: %d | ceiling: %d",
              n_ex4_cons, round(n_ex4_cons), ceiling(n_ex4_cons)))
```

**Interpretation (Exercise 4):** An overall interval width of $2\%$ corresponds to margin of error $E = 0.01$. Using the prior estimate $p = 0.85$ gives $4897.86$ (**$4{,}898$** consumers). Without prior information ($p = 0.50$), the conservative sample size would be **$9{,}604$** consumers.

Data Set: `hot_dogs.csv` (Consumer Reports, June 1986, pp. 366–367). Fifty-four observations across three hot dog types (`Beef`, `Meat`, `Poultry`), recording `Calories` and `Sodium`.

1. Create horizontal boxplots and find $95\%$ confidence intervals for the mean amount of calories in each `Type` of hot dog: `Beef`, `Meat`, and `Poultry`.
2. Construct $99\%$ one-sided lower confidence intervals for the mean amount of calories in each `Type` of hot dog: `Beef`, `Meat`, and `Poultry`.

```r
print(str(hotdogs))

boxplot(Calories ~ Type, data = hotdogs, horizontal = TRUE,
        col = c("salmon", "tan", "lightblue"),
        main = "Calories by Hot Dog Type",
        xlab = "Calories per Hot Dog", ylab = "Type")

beef <- subset(hotdogs, subset = (Type == "Beef"))
meat <- subset(hotdogs, subset = (Type == "Meat"))
poultry <- subset(hotdogs, subset = (Type == "Poultry"))

for (grp_name in c("Beef", "Meat", "Poultry")) {
  sub_df <- subset(hotdogs, Type == grp_name)
  ci95 <- t.test(sub_df$Calories, conf.level = 0.95)$conf.int
  ci99_less <- t.test(sub_df$Calories, alternative = "less", conf.level = 0.99)$conf.int
  ci99_greater <- t.test(sub_df$Calories, alternative = "greater", conf.level = 0.99)$conf.int
  print(sprintf("%s (n=%d, mean=%.2f) -> 95%% CI: [%.4f, %.4f]",
                grp_name, nrow(sub_df), mean(sub_df$Calories), ci95[1], ci95[2]))
  print(sprintf("   99%% one-sided (alternative='less'): (-Inf, %.4f] | (alternative='greater'): [%.4f, Inf)",
                ci99_less[2], ci99_greater[1]))
}
```

**Interpretation (Exercise 5):**

* **One-Sided $99\%$ CIs:** In R's `t.test()`, `alternative = "less"` produces an upper-bounded interval $(-\infty, U]$ (`169.71` for Beef, `174.52` for Meat, `132.90` for Poultry—as shown in the Lesson 08 Solutions PDF), while `alternative = "greater"` produces a lower-bounded interval $[L, +\infty)$ (`143.99` for Beef, `142.89` for Meat, `104.63` for Poultry).

## Exercise 6: $95\%$ Confidence Intervals for Mean and Variance of Hot Dog Calories (`hot_dogs.csv`)

Find a $95\%$ confidence interval for the mean and variance in the amount of calories found for each type of hotdog: `Beef`, `Meat`, and `Poultry`.

```r
var_conf_int <- function(x, conf_level = 0.95) {
  df <- length(x) - 1
  chilower <- qchisq((1 - conf_level) / 2, df, lower.tail = TRUE)
  chiupper <- qchisq((1 - conf_level) / 2, df, lower.tail = FALSE)
  v <- var(x)
  c(df * v / chiupper, df * v / chilower)
}

for (grp_name in c("Beef", "Meat", "Poultry")) {
  vals <- subset(hotdogs, Type == grp_name)$Calories
  mean_ci <- t.test(vals, conf.level = 0.95)$conf.int
  var_ci <- var_conf_int(vals, conf_level = 0.95)
  print(sprintf("%s -> Mean = %.2f (95%% CI: [%.4f, %.4f]) | Var = %.2f (95%% CI: [%.4f, %.4f])",
                grp_name, mean(vals), mean_ci[1], mean_ci[2], var(vals), var_ci[1], var_ci[2]))
}
```

**Interpretation (Exercise 6):** Across the three types, `Beef` ($s^2 = 512.66$, $95\%$ CI $[296.50, 1093.64]$), `Meat` ($s^2 = 636.85$, $95\%$ CI $[353.25, 1475.10]$), and `Poultry` ($s^2 = 508.57$, $95\%$ CI $[282.09, 1177.98]$) show substantial overlap in calorie variances even though `Poultry` has a noticeably lower mean calorie level. Remember that the $\chi^2$ variance confidence interval assumes each subgroup's calories are approximately normally distributed.
