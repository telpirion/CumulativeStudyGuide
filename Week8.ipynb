## Experimental design & one-way ANOVA — Questions 1–10

> **1.** A plan for testing hypotheses in which the researcher either controls or manipulates one or more variables is called a(n) \_\_\_\_\_\_\_.

* **Answer:** **`Experimental Design`**
* **Solution:**
  An experimental design is a structured plan for assigning treatments and controlling or manipulating variables to test hypotheses about population responses.

---

> **2.** A variable that is either controlled or manipulated is called a(n) \_\_\_\_\_\_\_ variable.

* **Answer:** **`Independent`**
* **Solution:**
  The independent variable is the variable deliberately controlled or varied by the researcher to observe its effect on the response.

---

> **3.** An independent variable is sometimes referred to as a \_\_\_\_\_\_\_ variable, a \_\_\_\_\_\_\_ variable, or a \_\_\_\_\_\_\_.

* **Answer:** **`Classification`**, **`Treatment`**, **`Factor`**
* **Solution:**
  In ANOVA terminology, an independent variable is commonly called a classification variable, a treatment variable, or a factor.

---

> **4.** Each independent variable contains two or more \_\_\_\_\_\_\_ or \_\_\_\_\_\_\_.

* **Answer:** **`Levels`**, **`Classifications`**
* **Solution:**
  The distinct categories or values of a factor are referred to as levels or classifications.

---

> **5.** The response to the different levels of the independent variables is called the \_\_\_\_\_\_\_ variable.

* **Answer:** **`Dependent`**
* **Solution:**
  The dependent (response) variable is the measured outcome whose variation is analyzed across factor levels.

---

> **6.** The experimental design that contains only one independent variable with two or more treatment levels is called a \_\_\_\_\_\_\_.

* **Answer:** **`Completely Randomized Design`**
* **Solution:**
  When subjects or units are assigned randomly to the treatment levels of a single factor, the design is a completely randomized design (one-way ANOVA).

---

> **7.** In chapter 11, the experimental designs are analyzed statistically using \_\_\_\_\_\_\_.

* **Answer:** **`Analysis of Variance (ANOVA)`**
* **Solution:**
  ANOVA compares group means simultaneously by partitioning total sample variance into between-group and within-group components.

---

> **8.** Suppose we want to analyze the data shown below using analysis of variance.

> **10.** For the problem presented in question 8, the sum of squares between is \_\_\_\_\_\_\_ and the sum of squares error is \_\_\_\_\_\_\_. The mean square between is \_\_\_\_\_\_\_ and the mean square error is \_\_\_\_\_\_\_. The observed value of F for this problem is \_\_\_\_\_\_\_. The decision is to \_\_\_\_\_\_\_.

* **Answer:** **`64.939`**, **`10.333`**, **`21.646`**, **`0.574`**, **`37.71`**, **`Reject the Null Hypothesis`**

## Multiple comparisons (Tukey's HSD & Tukey-Kramer) — Questions 11–15

> **11.** A set of techniques used to make comparisons between groups after an overall significant F value has been obtained is called \_\_\_\_\_\_\_.

* **Answer:** **`Multiple Comparisons`**
* **Solution:**
  Multiple comparison (post-hoc) procedures determine which specific pairs of population means differ significantly while controlling the family-wise Type I error rate.

---

> **12.** The two types of multiple comparison techniques presented in chapter 11 are \_\_\_\_\_\_\_ and \_\_\_\_\_\_\_.

* **Answer:** **`Tukey's Honestly Significant Difference Test (HSD)`**, **`Tukey-Kramer Procedure`**
* **Solution:**
  Tukey's HSD applies when all group sample sizes are equal, whereas the Tukey-Kramer procedure generalizes the test to unequal sample sizes.

---

> **13.** In conducting multiple comparisons with unequal sample sizes with techniques presented in chapter 11 of the text, you would use which procedure? \_\_\_\_\_\_\_

* **Answer:** **`Tukey-Kramer Procedure`**
* **Solution:**
  The Tukey-Kramer procedure uses the harmonic combination $\sqrt{\frac{MSE}{2}\left(\frac{1}{n_i} + \frac{1}{n_j}\right)}$ to accommodate unequal group sizes $n_i \ne n_j$.

---

> **14.** Suppose the following data are taken as samples from three populations and that an ANOVA results in an overall significant F value of 404.80. The mean square error for this ANOVA is 1.58.
>
> | Group 1 | Group 2 | Group 3 |
> | :---: | :---: | :---: |
> | 11 | 24 | 27 |
> | 9 | 25 | 30 |
> | 10 | 25 | 29 |
> | 12 | 26 | 28 |
> | 11 | 24 | 31 |
> | 8 | | 29 |
> | 10 | | |
>
> The Tukey-Kramer significant difference for groups 1 and 2 is \_\_\_\_\_\_\_. For groups 1 and 3, it is \_\_\_\_\_\_\_. For groups 2 and 3, it is \_\_\_\_\_\_\_. The following groups are significantly different \_\_\_\_\_\_\_ using $\alpha = .01$.

* **Answer:** **`2.514`**, **`2.388`**, **`2.60`**, **`All are significantly different`**
* **Solution:**
  With $C = 3$, $N = 7 + 5 + 6 = 18$, $df_{\text{err}} = 18 - 3 = 15$, and $\alpha = 0.01$, the Studentized range value is $q_{0.01, \, 3, \, 15} \approx 4.83$ (exact `qtukey(0.99, 3, 15) = 4.8359`, giving $2.517, 2.391, 2.603$):
  - Groups 1 & 2: $\text{CR}_{12} = q \sqrt{\frac{1.58}{2}\left(\frac{1}{7} + \frac{1}{5}\right)} \approx 2.514$, while $|\bar{x}_1 - \bar{x}_2| = |10.14 - 24.80| = 14.66 > 2.514$
  - Groups 1 & 3: $\text{CR}_{13} = q \sqrt{\frac{1.58}{2}\left(\frac{1}{7} + \frac{1}{6}\right)} \approx 2.388$, while $|\bar{x}_1 - \bar{x}_3| = |10.14 - 29.00| = 18.86 > 2.388$
  - Groups 2 & 3: $\text{CR}_{23} = q \sqrt{\frac{1.58}{2}\left(\frac{1}{5} + \frac{1}{6}\right)} \approx 2.60$, while $|\bar{x}_2 - \bar{x}_3| = |24.80 - 29.00| = 4.20 > 2.60$
  Every pairwise mean difference exceeds its critical range, so **all pairs are significantly different**.

> **15.** Suppose the following data represent four samples of size five which are taken from four populations. An ANOVA revealed a significant overall F value.
>
> | Group 1 | Group 2 | Group 3 | Group 4 |
> | :---: | :---: | :---: | :---: |
> | 5 | 11 | 12 | 21 |

## Randomized block design — Questions 16–20

> **16.** A research design that is similar to the completely randomized design except that it includes a second variable referred to as a blocking variable is called a(n) \_\_\_\_\_\_\_.

* **Answer:** **`Randomized Block Design`**
* **Solution:**
  A randomized block design groups experimental units into homogeneous blocks so that variability attributable to the blocking variable can be separated from error variance.

---

> **17.** In the randomized block design, the variable that the researcher desires to control but is not the treatment variable of interest is called the \_\_\_\_\_\_\_ variable.

* **Answer:** **`Blocking`**
* **Solution:**
  The blocking variable accounts for systematic nuisance variation across blocks, increasing the precision of the test for treatment effects.

---

> **18.** Consider the following randomized block design.
>
> | Block | Treatment 1 | Treatment 2 | Treatment 3 |
> | :---: | :---: | :---: | :---: |
> | 1 | 2 | 4 | 8 |
> | 2 | 3 | 4 | 9 |
> | 3 | 2 | 5 | 7 |
> | 4 | 4 | 6 | 6 |
> | 5 | 3 | 5 | 9 |
>
> The degrees of freedom treatment are \_\_\_\_\_\_\_. The degrees of freedom blocking are \_\_\_\_\_\_\_. The degrees of freedom error are \_\_\_\_\_\_\_.

* **Answer:** **`2`**, **`4`**, **`8`**
* **Solution:**
  With $k = 3$ treatment levels and $b = 5$ blocks:
  $$df_{\text{trt}} = k - 1 = 2, \qquad df_{\text{blk}} = b - 1 = 4, \qquad df_{\text{err}} = (k - 1)(b - 1) = 2 \times 4 = 8$$

---

> **19.** For the problem in question 18, the sum of squares treatment is \_\_\_\_\_\_\_. The sum of squares blocking are \_\_\_\_\_\_\_. The sum of squares error are \_\_\_\_\_\_\_.

* **Answer:** **`63.33`**, **`2.40`**, **`10.00`**
* **Solution:**
  The sum of all 15 observations is $77$, giving grand mean $\bar{\bar{x}} = \frac{77}{15} = 5.1333$. Treatment means are $\bar{x}_{\cdot 1} = 2.8$, $\bar{x}_{\cdot 2} = 4.8$, and $\bar{x}_{\cdot 3} = 7.8$:
  - $SS_{\text{trt}} = 5\big[(2.8 - 5.1333)^2 + (4.8 - 5.1333)^2 + (7.8 - 5.1333)^2\big] = 63.33$
  - Block means are $4.6667, 5.3333, 4.6667, 5.3333, 5.6667$, yielding $SS_{\text{blk}} = 2.40$
  - $SS_{\text{tot}} = 75.733 \implies SS_{\text{err}} = 75.733 - 63.333 - 2.40 = 10.00$

## Two-way factorial ANOVA — Questions 21–23

> **21.** One advantage of a two-way design over the completely randomized design and the randomized block design is that the researcher can test for \_\_\_\_\_\_\_ if multiple measures are taken under every combination of treatment levels of the two treatments.

* **Answer:** **`Interaction`**
* **Solution:**
  Replicating observations within every combination of factor levels permits separating interaction sum of squares from error, enabling a formal $F$-test for interaction effects.

---

## Correlation & simple linear regression concepts — Questions 1–14

> **1.** \_\_\_\_\_\_\_ is a measure of the degree of relatedness of two variables.

* **Answer:** **`Correlation`**
* **Solution:**

---

> **2.** The Pearson product-moment correlation coefficient is denoted by \_\_\_\_\_\_\_.

* **Answer:** **`r`**
* **Solution:**
  The sample Pearson correlation coefficient is denoted $r$ (and the population correlation parameter by $\rho$).

---

> **3.** The value of $r$ varies from \_\_\_\_\_\_\_.

* **Answer:** **`-1 to 0 to +1`**
* **Solution:**
  Correlation is bounded between $-1$ (perfect negative linear association) and $+1$ (perfect positive linear association), with $0$ indicating no linear correlation.

---

> **4.** Perfect positive correlation results in an $r$ value of \_\_\_\_\_\_\_.

* **Answer:** **`+1`**
* **Solution:**
  When all $(x_i, y_i)$ points fall exactly on a line with positive slope, $r = +1$.

---

> **5.** The value of the coefficient of correlation from the following data is \_\_\_\_\_\_\_.
>
> | $x$ | 19 | 20 | 26 | 31 | 34 | 45 | 45 | 51 |
> | :--- | :---: | :---: | :---: | :---: | :---: | :---: | :---: | :---: |
> | $y$ | 78 | 100 | 125 | 120 | 119 | 130 | 145 | 143 |

* **Answer:** **`0.876`** (`.876`)
* **Solution:**
  With $n = 8$, $\bar{x} = 33.875$, $\bar{y} = 120.0$, $SS_{xx} = 1004.875$, $SS_{yy} = 3424.0$, and $SS_{xy} = 1625.0$:
  $$r = \frac{SS_{xy}}{\sqrt{SS_{xx} \cdot SS_{yy}}} = \frac{1625.0}{\sqrt{1004.875 \times 3424.0}} \approx 0.876$$

---

> **6.** The value of $r$ from the following data is \_\_\_\_\_\_\_.

---

> **7.** The process of constructing a mathematical model or function that can be used to predict or determine one variable by another variable is \_\_\_\_\_\_\_.

* **Answer:** **`Regression`**

---

> **16.** A researcher wants to develop a regression line from the data given below:

> **17.** In regression, the value of $y - \hat{y}$ is called the \_\_\_\_\_\_\_.

* **Solution:**

> **18.** Data points that lie apart from the rest of the points are called \_\_\_\_\_\_\_.

* **Solution:**

> **19.** The regression assumption of constant error variance is called \_\_\_\_\_\_\_. If the error variances are not constant, it is called \_\_\_\_\_\_\_.

> **20.** Suppose the following data are used to determine the equation of the regression line given below:

> **21.** The total of the residuals squared is called the \_\_\_\_\_\_\_.

* **Answer:** **`Sum of Squares of Error`** ($SSE$)
* **Solution:**

## Standard error of the estimate & coefficient of determination — Questions 22–31

> **22.** A standard deviation of the error of the regression model is called the \_\_\_\_\_\_\_ and is denoted by \_\_\_\_\_\_\_.

* **Answer:** **`Standard Error of the Estimate`**, **`s_e`**
* **Solution:**
  The standard error of the estimate $s_e$ estimates the standard deviation of the error term around the regression line.

---

> **23.** Suppose a regression model is developed for ten pairs of data resulting in $\text{SSE} = 1,203$. The standard error of the estimate is \_\_\_\_\_\_\_.

* **Answer:** **`12.263`**
* **Solution:**
  With $n = 10$:
  $$s_e = \sqrt{\frac{SSE}{n - 2}} = \sqrt{\frac{1203}{10 - 2}} = \sqrt{150.375} \approx 12.263$$

---

> **24.** A regression analysis results in the following data:
> $$\sum x = 276, \quad \sum x^2 = 12,014, \quad \sum xy = 2,438, \quad \sum y = 77, \quad \sum y^2 = 1,183, \quad n = 7$$
> The value of $SSE$ is \_\_\_\_\_\_\_.

* **Answer:** **`20.015`** (exact `20.016`)
* **Solution:**
  First find the regression coefficients:
  $$SS_{xx} = 12014 - \frac{276^2}{7} = 1131.714, \quad SS_{xy} = 2438 - \frac{276(77)}{7} = -598$$
  $$b_1 = \frac{-598}{1131.714} \approx -0.5284, \quad b_0 = \frac{77}{7} - (-0.5284)\frac{276}{7} \approx 31.8341$$
  $$SSE = \sum y^2 - b_0 \sum y - b_1 \sum xy = 1183 - 31.8341(77) - (-0.5284)(2438) \approx 20.016$$

---

> **25.** The value of $s_e$ computed from the data of question 24 is \_\_\_\_\_\_\_.

* **Answer:** **`2.00`**
* **Solution:**
  $$s_e = \sqrt{\frac{SSE}{n - 2}} = \sqrt{\frac{20.016}{7 - 2}} = \sqrt{4.0032} \approx 2.00$$

> **26.** Suppose a regression model results in a value of $s_e = 27.9$. 95% of the residuals should fall within \_\_\_\_\_\_\_.

* **Answer:** **`0 ± 55.8`**
* **Solution:**
  By the Empirical Rule (for normally distributed residuals centered at 0), approximately 95% of residuals fall within $0 \pm 2s_e = 0 \pm 2(27.9) = 0 \pm 55.8$.

> **27.** Coefficient of determination is denoted by \_\_\_\_\_\_\_.

* **Answer:** **`r^2`** ($r^2$ or $R^2$)
* **Solution:**
  In simple linear regression, the coefficient of determination equals the square of the correlation coefficient, denoted $r^2$.

---

> **28.** \_\_\_\_\_\_\_ is the proportion of variability of the dependent variable accounted for or explained by the independent variable.

* **Answer:** **`Coefficient of Determination`**
* **Solution:**
  $r^2 = \frac{SSR}{SST} = 1 - \frac{SSE}{SST}$ gives the proportion of total variation in $Y$ explained by $X$.

---

> **29.** The value of $r^2$ always falls between \_\_\_\_\_\_\_ and \_\_\_\_\_\_\_ inclusive.

* **Answer:** **`0`**, **`1`**
* **Solution:**
  Because $0 \le SSR \le SST$, the ratio $r^2 = SSR / SST$ is always bounded between $0$ and $1$.

---

> **30.** Suppose a regression analysis results in the following:
> $$b_1 = .19364, \quad b_0 = 59.4798, \quad n = 8, \quad \sum y = 1,019, \quad \sum y^2 = 134,451, \quad \sum xy = 378,932$$
> The value of $r^2$ for this regression model is \_\_\_\_\_\_\_.

* **Answer:** **`0.900`** (`.900`)
* **Solution:**
  $$SSE = \sum y^2 - b_0 \sum y - b_1 \sum xy = 134451 - 59.4798(1019) - 0.19364(378932) \approx 464.71$$
  $$SST = \sum y^2 - \frac{(\sum y)^2}{n} = 134451 - \frac{1019^2}{8} = 4655.875$$
  $$r^2 = 1 - \frac{SSE}{SST} = 1 - \frac{464.71}{4655.875} \approx 0.900$$

**Dataset:** [`tableware.csv`](https://raw.githubusercontent.com/telpirion/CumulativeStudyGuide/refs/heads/main/data/tableware.csv)

```r
anova_rate <- aov(RATE ~ TYPE, data = tableware)
print(summary(anova_rate))
boxplot(RATE ~ TYPE, data = tableware, horizontal = TRUE, main = "RATE by TYPE",
        xlab = "Rate", ylab = "Type")

anova_price <- aov(PRICE ~ TYPE, data = tableware)
print(summary(anova_price))
if (summary(anova_price)[[1]][["Pr(>F)"]][1] < 0.05) {
  print(TukeyHSD(anova_price))
}
boxplot(PRICE ~ TYPE, data = tableware, horizontal = TRUE, main = "PRICE by TYPE",
        xlab = "Price", ylab = "Type")
```

We reject $H_0$; mean `PRICE` differs significantly by tableware type, with Tukey's HSD identifying significant pairwise differences between `cass` and `bowl` ($p_{\text{adj}} = 0.0006$), `plate` and `cass` ($p_{\text{adj}} = 0.0003$), and `tray` and `plate` ($p_{\text{adj}} = 0.0499$).
