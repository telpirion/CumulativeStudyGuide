## Tableware Data Analysis
We are testing if the mean `RATE` and `PRICE` for the five levels of `TYPE` are equal.


```python
print("R Code for Tableware Analysis")
print("tableware <- read.csv('data/tableware.csv')")
print("anova_rate <- aov(RATE ~ TYPE, data = tableware)")
print("print(summary(anova_rate))")
print("boxplot(RATE ~ TYPE, data = tableware, horizontal = TRUE, main = 'RATE by TYPE')")
print("anova_price <- aov(PRICE ~ TYPE, data = tableware)")
print("print(summary(anova_price))")
print("boxplot(PRICE ~ TYPE, data = tableware, horizontal = TRUE, main = 'PRICE by TYPE')")
```

## Hot Dogs Data Analysis
We will perform one-way ANOVA by `Type` on `Calories` and `Sodium`, followed by Tukey's HSD.


```python
print("R Code for Hot Dogs Analysis")
print("hot_dogs <- read.csv('data/hot_dogs.csv')")
print("anova_cal <- aov(Calories ~ Type, data = hot_dogs)")
print("print(summary(anova_cal))")
print("if (summary(anova_cal)[[1]][['Pr(>F)']][1] < 0.05) print(TukeyHSD(anova_cal))")
print("boxplot(Calories ~ Type, data = hot_dogs, horizontal = TRUE, main = 'Calories by Type')")
print("anova_sod <- aov(Sodium ~ Type, data = hot_dogs)")
print("print(summary(anova_sod))")
print("if (summary(anova_sod)[[1]][['Pr(>F)']][1] < 0.05) print(TukeyHSD(anova_sod))")
print("boxplot(Sodium ~ Type, data = hot_dogs, horizontal = TRUE, main = 'Sodium by Type')")
```
