# Chapter 4 study questions (Probability)

The following 41 questions provide applied review across all core topics of Chapter 4, including methods of assigning probability, counting rules, set operations, independence, mutually exclusive events, contingency tables, and Bayes' Theorem. *(Note: In the original Chapter 4 practice PDF, question numbers 3 and 4 were duplicated/skipped in the prompt; all 41 items are included below in sequential order).*

## Methods of assigning probability & basic concepts — Questions 1–5

> **1.** The \_\_\_\_\_\_\_ method of assigning probabilities relies on the insight or feelings of the person determining the probabilities.

* **Answer:** **`Subjective`**
* **Solution:**
  Subjective probability relies on personal judgment, intuition, expertise, or feelings when historical data or equally likely classical outcomes are unavailable.

---

> **2.** If probabilities are determined "a priori" to an experiment using rules and laws, then the \_\_\_\_\_\_\_ method of assigning probabilities is being used.

* **Answer:** **`Classical`**
* **Solution:**
  Classical probability determines likelihoods *a priori* (before running the experiment) based on logical symmetry and equally likely elementary outcomes: $P(E) = \frac{n_E}{N}$.

---

> **3.** The range of possibilities for probability values is from \_\_\_\_\_\_\_ to \_\_\_\_\_\_\_.

* **Answer:** **`0` to `1`** (inclusive)
* **Solution:**
  Every valid probability satisfies $0 \le P(E) \le 1$, where $0$ indicates impossibility and $1$ indicates certainty.

---

> **4.** *(Numbered 3 in original PDF)* Suppose a technician keeps track of all defects in raw materials for a single day and uses this information to determine the probability of finding a defect in raw materials the next day. She is using the \_\_\_\_\_\_\_ method of assigning probabilities.

* **Answer:** **`Relative Frequency`**
* **Solution:**
  Empirical probability computed from observed historical frequencies divided by total trials ($n_E / N$) is the relative frequency of occurrence method.

---

> **5.** *(Numbered 4 in original PDF)* The outcome of an experiment is called a(n) \_\_\_\_\_\_\_. If these outcomes cannot be decomposed further, then they are referred to as \_\_\_\_\_\_\_.

* **Answer:** **`Event`**, **`Elementary Events`**
* **Solution:**
  An **event** is any outcome or subset of outcomes of an experiment; **elementary events** are the indivisible basic outcomes that form the sample space.

## Counting rules, sample spaces & set operations — Questions 6–9

> **6.** A computer hardware retailer allows you to order your own computer monitor. The store carries five different brands of monitors. Each brand comes in 14", 15" or 17" models. In addition, you can purchase either the deluxe model or the regular model in each brand and in each size. How many different types of monitors are available considering all the factors? \_\_\_\_\_\_\_ You probably used the \_\_\_\_\_\_\_ rule to solve this.

* **Answer:** **`30`**, **`mn counting rule`**
* **Solution:**
  Multiplying the number of options across the three factors: $5 \times 3 \times 2 = 30$ distinct monitor configurations via the $mn$ counting rule.

---

> **7.** Suppose you are playing the Lotto game and you are trying to "pick" three numbers. For each of the three numbers, any of the digits 0 through 9 are possible (with replacement). How many different sets of numbers are available? \_\_\_\_\_\_\_

* **Answer:** **`10^3 = 1000` numbers**
* **Solution:**
  Each of the 3 positions has 10 independent choices ($0$–$9$) with replacement: $10 \times 10 \times 10 = 10^3 = 1{,}000$.

---

> **8.** A population consists of the odd numbers between 1 and 9 inclusive. If a researcher randomly samples numbers from the population three at a time, the sample space is \_\_\_\_\_\_\_. Using combinations, how could we have determined ahead of time how many elementary events would be in the sample space? \_\_\_\_\_\_\_

* **Answer:** **`{(1,3,5), (1,3,7), (1,3,9), (1,5,7), (1,5,9), (1,7,9), (3,5,7), (3,5,9), (3,7,9), (5,7,9)}`**, **`5C3 = 10`**
* **Solution:**
  The population is $\{1, 3, 5, 7, 9\}$ ($N = 5$). Sampling $n = 3$ without replacement yields $\binom{5}{3} = \frac{5!}{3!(5-3)!} = \frac{120}{6 \cdot 2} = 10$ combinations.

---

> **9.** Let $A = \{2, 3, 5, 6, 7, 9\}$ and $B = \{1, 3, 4, 6, 7, 9\}$.
> $A \cup B =$ \_\_\_\_\_\_\_ and $A \cap B =$ \_\_\_\_\_\_\_.

* **Answer:** **`{1, 2, 3, 4, 5, 6, 7, 9}`**, **`{3, 6, 7, 9}`**
* **Solution:**
  - The union $A \cup B$ collects all distinct elements in either $A$ or $B$: $\{1, 2, 3, 4, 5, 6, 7, 9\}$.
  - The intersection $A \cap B$ collects only elements common to both sets: $\{3, 6, 7, 9\}$.

## Independence, mutually exclusive events & complements — Questions 10–15

> **10.** If the occurrence of one event does not affect the occurrence of the other event, then the events are said to be \_\_\_\_\_\_\_.

* **Answer:** **`Independent`**
* **Solution:**
  Events $A$ and $B$ are statistically independent when $P(A \mid B) = P(A)$ and $P(B \mid A) = P(B)$.

---

> **11.** The outcome of the roll of one die is said to be \_\_\_\_\_\_\_ of the outcome of the roll of another die.

* **Answer:** **`Independent`**
* **Solution:**
  Rolling one fair die does not physically or probabilistically influence the outcome of rolling a second die.

---

> **12.** The event of rolling a three on a die and the event of rolling an even number on the same roll with the same die are \_\_\_\_\_\_\_.

* **Answer:** **`Mutually Exclusive`**
* **Solution:**
  A single roll cannot simultaneously be $3$ (odd) and an even number ($\{2, 4, 6\}$), so their intersection is empty.

---

> **13.** If the probability of the intersection of two events is zero, then the events are said to be \_\_\_\_\_\_\_.

* **Answer:** **`Mutually Exclusive`**
* **Solution:**
  $P(A \cap B) = 0$ defines mutually exclusive (disjoint) events that cannot occur together.

---

> **14.** If three objects are selected from a bin, one at a time with replacement, the outcomes of each selection are \_\_\_\_\_\_\_.

* **Answer:** **`Independent`**
* **Solution:**
  Because each selected object is returned to the bin before the next draw, the composition of the bin—and thus all probabilities—remains unchanged across draws.

---

> **15.** Suppose a population consists of a manufacturing facility's 1600 workers. Suppose an experiment is conducted in which a worker is randomly selected. If an event is the selection of a worker over 40 years old, then the event of selecting a worker 40 years or younger is called the \_\_\_\_\_\_\_ of the first event.

* **Answer:** **`Complement`**
* **Solution:**
  "40 years or younger" contains every elementary outcome in the sample space not in "over 40 years old," forming the complement $A'$.

## Types of probability & core probability laws — Questions 16–23

> **16.** The probability of selecting $X$ given that $Y$ has occurred is called a \_\_\_\_\_\_\_ probability.

* **Answer:** **`Conditional`**
* **Solution:**
  Denoted $P(X \mid Y) = \frac{P(X \cap Y)}{P(Y)}$.

---

> **17.** The probability of $X$ is called a \_\_\_\_\_\_\_ probability.

* **Answer:** **`Marginal`**
* **Solution:**
  In a joint probability table, the unconditional probability of a single event $P(X)$ sits in the row or column margin.

---

> **18.** The probability of $X$ or $Y$ occurring is called a \_\_\_\_\_\_\_ probability.

* **Answer:** **`Union`**
* **Solution:**
  Denoted $P(X \cup Y)$, computed via the Addition Law.

---

> **19.** The probability of $X$ and $Y$ occurring is called a \_\_\_\_\_\_\_ probability.

* **Answer:** **`Joint`** (or **`Intersection`**)
* **Solution:**
  Denoted $P(X \cap Y)$, representing the simultaneous occurrence of both events.

---

> **20.** Only one of the four types of probability does not use the total possible outcomes in the denominator when calculating the probability. This type of probability is called \_\_\_\_\_\_\_ probability.

* **Answer:** **`Conditional`**
* **Solution:**
  Conditional probability $P(X \mid Y)$ restricts the sample space to the conditioning event $Y$, placing $P(Y)$ (or $n_Y$) in the denominator instead of the grand total $N$.

---

> **21.** If the $P(A \mid B) = P(A)$, then the events $A, B$ are \_\_\_\_\_\_\_ events.

* **Answer:** **`Independent`**
* **Solution:**
  Knowing $B$ occurred does not alter the likelihood of $A$, which is the formal definition of independence.

---

> **22.** If the $P(X) = .53$, the $P(Y) = .12$, and the $P(X \cap Y) = .07$, then $P(X \cup Y) =$ \_\_\_\_\_\_\_.

* **Answer:** **`0.58`** (`.58`)
* **Solution:**
  By the General Law of Addition:
  $$P(X \cup Y) = P(X) + P(Y) - P(X \cap Y) = 0.53 + 0.12 - 0.07 = 0.58$$

---

> **23.** If the $P(X) = .26$, the $P(Y) = .31$, and $X, Y$ are mutually exclusive, then $P(X \cup Y) =$ \_\_\_\_\_\_\_.

* **Answer:** **`0.57`** (`.57`)
* **Solution:**
  Because $X$ and $Y$ are mutually exclusive, $P(X \cap Y) = 0$. By the Special Law of Addition:
  $$P(X \cup Y) = P(X) + P(Y) = 0.26 + 0.31 = 0.57$$

## Probability matrices, multiplication & conditional laws — Questions 24–27

> **24.** In a company, 47% of the employees wear glasses, 60% of the employees are women, and 28% of the employees are women and wear glasses. Complete the probability matrix below for this problem.

* **Answer:**

| Gender | Wear Glasses: Yes | Wear Glasses: No | Total |
| :--- | :---: | :---: | :---: |
| **Men** | **0.19** | **0.21** | **0.40** |
| **Women** | **0.28** | **0.32** | **0.60** |
| **Total** | **0.47** | **0.53** | **1.00** |

* **Solution:**
  - Given: $P(\text{Yes}) = 0.47$, $P(\text{Women}) = 0.60$, $P(\text{Women} \cap \text{Yes}) = 0.28$.
  - Complements: $P(\text{Men}) = 1 - 0.60 = 0.40$; $P(\text{No}) = 1 - 0.47 = 0.53$.
  - Remaining cells by row/column subtraction:
    * $P(\text{Men} \cap \text{Yes}) = 0.47 - 0.28 = 0.19$
    * $P(\text{Women} \cap \text{No}) = 0.60 - 0.28 = 0.32$
    * $P(\text{Men} \cap \text{No}) = 0.40 - 0.19 = 0.21$ (check: $0.21 + 0.32 = 0.53$).

---

> **25.** Suppose that in another company, 40% of the workers are part time and 80% of the part time workers are men. The probability of randomly selecting a company worker who is both part time and a man is \_\_\_\_\_\_\_.

* **Answer:** **`0.32`** (`.32`)
* **Solution:**
  By the General Law of Multiplication:
  $$P(\text{PT} \cap \text{Man}) = P(\text{PT}) \cdot P(\text{Man} \mid \text{PT}) = 0.40 \times 0.80 = 0.32$$

---

> **26.** The probability of tossing three coins in a row and getting all tails is \_\_\_\_\_\_\_. This is an application of the \_\_\_\_\_\_\_ law of multiplication because each toss is \_\_\_\_\_\_\_.

* **Answer:** **`1/8 = 0.125`**, **`Special`**, **`Independent`**
* **Solution:**
  Since coin tosses are independent, the Special Law of Multiplication applies:
  $$P(T_1 \cap T_2 \cap T_3) = (0.5)(0.5)(0.5) = \frac{1}{8} = 0.125$$

---

> **27.** Suppose 70% of all cars purchased in America are U.S.A. made and that 18% of all cars purchased in America are both U.S.A. made and are red. The probability that a randomly selected car purchased in America is red given that it is U.S.A. made is \_\_\_\_\_\_\_.

* **Answer:** **`0.2571`** (`.2571`)
* **Solution:**
  $$P(\text{Red} \mid \text{U.S.A.}) = \frac{P(\text{Red} \cap \text{U.S.A.})}{P(\text{U.S.A.})} = \frac{0.18}{0.70} \approx 0.2571$$

## Conditional, union & Bayes' theorem problems — Questions 38–41

> **38.** Suppose 42% of all people in a county have characteristic $X$. Suppose 17% of all people in this county have characteristic $X$ and characteristic $Y$. If a person is randomly selected from the county who is known to have characteristic $X$, then the probability that they have characteristic $Y$ is \_\_\_\_\_\_\_.

* **Answer:** **`0.4048`** (`.4048`)
* **Solution:**
  $$P(Y \mid X) = \frac{P(X \cap Y)}{P(X)} = \frac{0.17}{0.42} \approx 0.4048$$

---

> **39.** Suppose 22% of all parts produced at a plant have flaw $X$ and 37% have flaw $Y$. In addition, suppose 53% of the parts with flaw $X$ have flaw $Y$. If a part is randomly selected, the probability that it has flaw $X$ or flaw $Y$ is \_\_\_\_\_\_\_.

* **Answer:** **`0.4734`** (`.4734`)
* **Solution:**
  First compute the joint probability: $P(X \cap Y) = P(X) \cdot P(Y \mid X) = 0.22 \times 0.53 = 0.1166$.
  Then apply the General Law of Addition:
  $$P(X \cup Y) = P(X) + P(Y) - P(X \cap Y) = 0.22 + 0.37 - 0.1166 = 0.4734$$

---

> **40.** Another name for revision of probabilities is \_\_\_\_\_\_\_.

* **Answer:** **`Bayes' Rule`** (or Bayes' Theorem)
* **Solution:**
  Bayes' Rule revises prior probabilities into posterior probabilities when new conditional evidence is observed.

---

> **41.** Suppose the prior probabilities of $A$ and $B$ are .57 and .43 respectively. Suppose that $P(E \mid A) = .24$ and $P(E \mid B) = .56$. If $E$ is known to have occurred, then the revised probability of $A$ occurring is \_\_\_\_\_\_\_ and of $B$ occurring is \_\_\_\_\_\_\_.

* **Answer:** **`0.3623`**, **`0.6377`**
* **Solution:**
  Compute the products $P(H_i)P(E \mid H_i)$ and total probability $P(E)$:
  - $P(A)P(E \mid A) = 0.57 \times 0.24 = 0.1368$
  - $P(B)P(E \mid B) = 0.43 \times 0.56 = 0.2408$
  - $P(E) = 0.1368 + 0.2408 = 0.3776$
  Posterior probabilities:
  $$P(A \mid E) = \frac{0.1368}{0.3776} \approx 0.3623, \qquad P(B \mid E) = \frac{0.2408}{0.3776} \approx 0.6377$$

## Random variables & discrete distribution summary — Questions 1–10

> **1.** Variables that take on values at every point over a given interval are called \_\_\_\_\_\_\_ \_\_\_\_\_\_\_ variables.

* **Answer:** **`Continuous Random`**
* **Solution:**
  Continuous random variables arise from measurement and can assume any real value along an interval.

---

> **2.** If the set of all possible values of a variable is at most finite or a countably infinite number of possible values, then the variable is called a \_\_\_\_\_\_\_ \_\_\_\_\_\_\_ variable.

* **Answer:** **`Discrete Random`**
* **Solution:**
  Discrete random variables take on countable, distinct values (often integers from counting).

---

> **3.** An experiment in which a die is rolled six times will likely produce values of a \_\_\_\_\_\_\_ random variable.

* **Answer:** **`Discrete`**
* **Solution:**
  Die outcomes ($1, 2, 3, 4, 5, 6$) and counts of specific faces are finite and countable.

---

> **4.** An experiment in which a researcher counts the number of customers arriving at a supermarket checkout counter every two minutes produces values of a \_\_\_\_\_\_\_ random variable.

* **Answer:** **`Discrete`**
* **Solution:**

> **5.** An experiment in which the time it takes to assemble a product is measured is likely to produce values of a \_\_\_\_\_\_\_ random variable.

---

> **6.** A binomial distribution is an example of a \_\_\_\_\_\_\_ distribution.

* **Answer:** **`Discrete`**
* **Solution:**
  The binomial random variable counts the integer number of successes $x \in \{0, 1, \dots, n\}$.

---

> **7.** The normal distribution is an example of a \_\_\_\_\_\_\_ distribution.

* **Answer:** **`Continuous`**
* **Solution:**

* **Answer:** **`Mean`**, **`Expected Value`**
* **Solution:**

## The binomial distribution — Questions 11–17

> **11.** On any one trial of a binomial experiment, there can be only \_\_\_\_\_\_\_ possible outcomes.

* **Answer:** **`Two`**
* **Solution:**
  Each Bernoulli trial results in one of two mutually exclusive outcomes: Success or Failure.

---

> **12.** Suppose the probability that a given part is defective is .10. If four such parts are randomly drawn from a large population, the probability that exactly two parts are defective is \_\_\_\_\_\_\_.

* **Answer:** **`0.0486`** (`.0486`)
* **Solution:**
  With $n = 4$, $p = 0.10$, and $x = 2$:
  $$P(X = 2) = \binom{4}{2}(0.10)^2(0.90)^2 = 6(0.01)(0.81) = 0.0486$$

---

> **13.** Suppose the probability that a given part is defective is .04. If thirteen such parts are randomly drawn from a large population, the expected value or mean of the binomial distribution that describes this experiment is \_\_\_\_\_\_\_.

* **Answer:** **`0.52`**
* **Solution:**
  $$\mu = n p = 13 \times 0.04 = 0.52$$

---

> **14.** Suppose a binomial experiment is conducted by randomly selecting 20 items where $p = .30$. The standard deviation of the binomial distribution is \_\_\_\_\_\_\_.

* **Answer:** **`2.049`**
* **Solution:**
  $$\sigma = \sqrt{n p (1 - p)} = \sqrt{20(0.30)(0.70)} = \sqrt{4.2} \approx 2.0494$$

---

> **15.** Suppose forty-seven percent of the workers in a large corporation are under thirty-five years of age. If fifteen workers are randomly selected from this corporation, the probability of selecting exactly ten who are under thirty-five years of age is \_\_\_\_\_\_\_.

* **Answer:** **`0.0661`** (`.0661`)
* **Solution:**
  With $n = 15$, $p = 0.47$, and $x = 10$:
  $$P(X = 10) = \binom{15}{10}(0.47)^{10}(0.53)^5 = 3003(0.47)^{10}(0.53)^5 \approx 0.0661$$

---

> **16.** Suppose that twenty-three percent of all adult Americans fly at least once a year. If twelve adult Americans are randomly selected, the probability that exactly four have flown at least once last year is \_\_\_\_\_\_\_.

* **Answer:** **`0.1712`** (`.1712`)
* **Solution:**
  With $n = 12$, $p = 0.23$, and $x = 4$:
  $$P(X = 4) = \binom{12}{4}(0.23)^4(0.77)^8 = 495(0.23)^4(0.77)^8 \approx 0.1712$$

---

> **17.** Suppose that sixty percent of all voters support the President of the United States at this time. If twenty voters are randomly selected, the probability that at least eleven support the President is \_\_\_\_\_\_\_.

* **Answer:** **`0.755`** (`.755`, exact `0.7553`)
* **Solution:**
  With $n = 20$ and $p = 0.60$:

### Exercise 1: Shopper Spending Probability Analysis

**Dataset:** [`shoppers.csv`](https://raw.githubusercontent.com/telpirion/CumulativeStudyGuide/refs/heads/main/data/shoppers.csv) contains the dollar amounts spent in a store by 50 individual shoppers during one day (`Spending`).

Assume the fifty shoppers exit the store individually in random order.

* **a)** If one shopper is picked at random, what is the probability of picking a shopper who spent \$40 or more dollars? What is the probability of picking a shopper who spent less than \$10?
* **b)** If two shoppers are picked at random, what is the probability the pair will include a shopper who spent \$40 or more dollars and one who spent less than \$10?
* **c)** *(Sampling without replacement)* If two shoppers are picked at random, what is the probability the pair will include two shoppers who spent no less than \$10 and no more than \$40?
* **d)** *(Sampling without replacement)* If four shoppers are picked at random, what is the probability one shopper will have spent less than \$10, one shopper will have spent \$40 or more dollars, and two shoppers will have spent no less than \$10 and no more than \$40?
* **e)** If we know a randomly picked shopper has spent more than \$30, what is the probability that shopper has spent more than \$40?

```r
print(str(shoppers))
print(table(shoppers$Spending >= 40))

n_total <- nrow(shoppers)
n_lt10 <- sum(shoppers$Spending < 10)
n_ge40 <- sum(shoppers$Spending >= 40)
n_mid <- sum(shoppers$Spending >= 10 & shoppers$Spending <= 40)

p_1a_ge40 <- n_ge40 / n_total
p_1a_lt10 <- n_lt10 / n_total
print(sprintf("1a) P(Spending >= $40) = %.4f | P(Spending < $10) = %.4f", p_1a_ge40, p_1a_lt10))

p_1b <- (n_lt10 * n_ge40) / (n_total * (n_total - 1) / 2)
print(sprintf("1b) P(one >= $40 and one < $10) = %.6f", p_1b))

p_1c <- (n_mid * (n_mid - 1)) / (n_total * (n_total - 1))
print(sprintf("1c) P(both in [$10, $40]) = %.6f", p_1c))

successful_4 <- n_lt10 * n_ge40 * (n_mid * (n_mid - 1) / 2)
total_4 <- choose(n_total, 4)
p_1d <- successful_4 / total_4
print(sprintf("1d) P(1 <$10, 1 >=$40, 2 in [$10, $40]) = %.6f", p_1d))

shoppers_gt30 <- subset(shoppers, subset = Spending > 30)
p_1e <- sum(shoppers_gt30$Spending > 40) / nrow(shoppers_gt30)
print(sprintf("1e) P(Spending > $40 | Spending > $30) = %.6f", p_1e))
```

**Conclusions for Exercise 1:**

* Out of $n = 50$ shoppers, $n_1 = 5$ spent under \$10 ($P = 0.10$), $n_2 = 8$ spent \$40 or more ($P = 0.16$), and $m = 37$ spent between \$10 and \$40 inclusive.
* Sampling without replacement adjusts denominators to $\binom{50}{2} = 1{,}225$ for pairs (parts b and c) and $\binom{50}{4} = 230{,}300$ for groups of four (part d).

### Exercise 2: Monte Carlo Probability Simulations in R

* **a) The Birthday Problem:** Draw 100 samples with replacement of size 22 from the 365 integers (`1:365`). Count the number of samples in which one or more of the numbers sampled is duplicated, and divide by 100 to estimate the probability of matching birthdays. Also refine the estimate using 10,000 replications.
* **b) Marble / Binomial Simulation:** Suppose that 60% of marbles in a bag are black and 40% are white. Generate a random sample of size 20 with replacement using uniform random numbers (`runif()`). Code numbers $\le 0.6$ as `1` and $> 0.6$ as `0`, then sum the 20 coded numbers. Repeat 50 times and calculate the proportion of times the sum is 11 or greater. Expand the number of trials to 10,000, plot a histogram, compute the expected value, and compare the simulated relative frequencies against the exact binomial distribution ($n = 20, p = 0.60$, where exact $P(X \ge 11) = 0.7553$ and $E(X) = 12$).

```r
set.seed(1234)
count_duplicates <- 0
for (i in 1:100) {
  this_sample <- sample(1:365, size = 22, replace = TRUE)
  if (length(this_sample) != length(unique(this_sample))) {
    count_duplicates <- count_duplicates + 1
  }
}
prob_any_duplicates <- count_duplicates / 100

set.seed(1234)
prob_rep_100 <- mean(replicate(100, any(duplicated(sample(1:365, 22, replace = TRUE)))))
set.seed(1234)
prob_rep_10000 <- mean(replicate(10000, any(duplicated(sample(1:365, 22, replace = TRUE)))))
exact_birthday <- 1 - prod(365:(365 - 21) / 365)

print(sprintf("2a) Loop (100): %.2f | Replicate (100): %.4f | Replicate (10,000): %.4f | Exact: %.4f",
              prob_any_duplicates, prob_rep_100, prob_rep_10000, exact_birthday))
```

```r
count_black <- function(N, p) {
  x <- runif(n = N)
  sum(x <= p)
}

set.seed(1234)
res_50 <- replicate(50, count_black(20, 0.6))
prop_50 <- mean(res_50 >= 11)

set.seed(1234)
res_10000 <- replicate(10000, count_black(20, 0.6))
mean_10000 <- mean(res_10000)
prop_10000 <- mean(res_10000 >= 11)
exact_binom_ge11 <- pbinom(10, size = 20, prob = 0.6, lower.tail = FALSE)

print(sprintf("2b) 50 trials P(X >= 11): %.4f", prop_50))
print(sprintf("2b) 10,000 trials Mean: %.4f | P(X >= 11): %.4f | Exact Binomial P(X >= 11): %.4f",
              mean_10000, prop_10000, exact_binom_ge11))

hist(res_10000, col = "lightblue", xlab = "Number of Black Marbles (out of 20)",
     main = "Histogram of 10,000 Binomial Simulation Trials")

outcome <- as.data.frame(table(res_10000))
relative_frequency <- outcome[, 2] / sum(outcome[, 2])
successes <- factor(outcome[, 1])
obs_x <- as.numeric(as.character(outcome[, 1]))
binomial_probabilities <- dbinom(obs_x, size = 20, prob = 0.6)

par(mfrow = c(1, 2))
barplot(binomial_probabilities, names.arg = successes, xlab = "Successes",
        ylab = "Binomial Probabilities", main = "Exact Binomial PMF", col = "steelblue")
barplot(relative_frequency, names.arg = successes, xlab = "Successes",
        ylab = "Relative Frequency", main = "Simulated (10,000 Trials)", col = "darkorange")
```

### Exercise 3: Bayesian Sequential Updating (*Think Bayes* Ch. 2–3)

1. **Cookie Problem Without Replacement (*Think Bayes* Exercise 2.1):** Bowl 1 begins with 30 vanilla and 10 chocolate cookies, and Bowl 2 begins with 20 vanilla and 20 chocolate cookies. Suppose you pick a bowl at random and draw **three cookies without replacement** (eating each cookie after drawing it): `vanilla`, `chocolate`, `vanilla`. Update the posterior probability that the cookies came from Bowl 1 after each draw, and contrast this with sampling *with* replacement.
2. **The Dice Problem (*Think Bayes* Section 3.1):** A box contains five dice with `4, 6, 8, 12, 20` sides. You pick one die at random and roll the sequence `6, 8, 7, 7, 5, 4`. Compute the posterior probability for each die after the first roll (`6`) and after all six rolls.

```r
run_cookie_without_replacement <- function(draws) {
  state <- list(
    Bowl1 = c(vanilla = 30, chocolate = 10),
    Bowl2 = c(vanilla = 20, chocolate = 20)
  )
  post <- c(Bowl1 = 0.5, Bowl2 = 0.5)
  for (i in seq_along(draws)) {
    d <- draws[i]
    like <- c(
      Bowl1 = state$Bowl1[[d]] / sum(state$Bowl1),
      Bowl2 = state$Bowl2[[d]] / sum(state$Bowl2)
    )
    post <- (post * like) / sum(post * like)
    state$Bowl1[[d]] <- state$Bowl1[[d]] - 1
    state$Bowl2[[d]] <- state$Bowl2[[d]] - 1
    print(sprintf("Draw %d (%s) -> P(Bowl 1): %.4f | P(Bowl 2): %.4f",
                  i, d, post["Bowl1"], post["Bowl2"]))
  }
}

run_cookie_without_replacement(c("vanilla", "chocolate", "vanilla"))

dice_sides <- c(4, 6, 8, 12, 20)
dice_post <- rep(1 / length(dice_sides), length(dice_sides))
names(dice_post) <- paste0("d", dice_sides)
rolls <- c(6, 8, 7, 7, 5, 4)

for (i in seq_along(rolls)) {
  r <- rolls[i]
  like <- ifelse(dice_sides < r, 0, 1 / dice_sides)
  dice_post <- (dice_post * like) / sum(dice_post * like)
  if (i == 1 || i == length(rolls)) {
    print(sprintf("After roll %d (value = %d):", i, r))
    print(round(dice_post, 4))
  }
}
```
