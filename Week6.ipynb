## Fundamentals of hypothesis testing & error types — Questions 1–14

> **1.** The first step in testing a hypothesis is to establish a(n) \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ hypothesis and a(n) \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ hypothesis.

* **Answer:** **`Null`**, **`Alternative`**
* **Solution:**
  Hypothesis testing begins by specifying the baseline **null hypothesis ($H_0$)** and the competing **alternative hypothesis ($H_a$)**.

---

> **2.** In testing hypotheses, the researcher initially assumes that the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ hypothesis is true.

* **Answer:** **`Null`**
* **Solution:**
  Inferential test statistics and $p$-values are computed under the assumption that the **null hypothesis ($H_0$)** holds true until sample evidence justifies rejecting it.

---

> **3.** The region of the distribution in hypothesis testing in which the null hypothesis is rejected is called the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ region.

* **Answer:** **`Rejection`**
* **Solution:**
  If the observed test statistic falls into the **rejection region** (beyond the critical value), $H_0$ is rejected.

---

> **4.** The rejection and acceptance regions are divided by a point called the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ value.

* **Answer:** **`Critical`**
* **Solution:**
  The **critical value** marks the boundary between the rejection region and the nonrejection region for a chosen significance level $\alpha$.

---

> **5.** The portion of the distribution which is not in the rejection region is called the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ region.

* **Answer:** **`Nonrejection Region`** (or acceptance region)
* **Solution:**
  Values of the test statistic falling inside the **nonrejection region** do not provide sufficient evidence to reject $H_0$.

---

> **6.** The probability of committing a Type I error is called \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`Alpha`** ($\alpha$)
* **Solution:**
  $\alpha = P(\text{Reject } H_0 \mid H_0 \text{ is true})$ is the probability of committing a Type I error.

---

> **7.** Another name for alpha is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`Level of Significance`**
* **Solution:**
  $\alpha$ is formally called the **level of significance** of the hypothesis test.

---

> **8.** When a true null hypothesis is rejected, the researcher has committed a \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ error.

* **Answer:** **`Type I`**
* **Solution:**
  Rejecting $H_0$ when $H_0$ is actually true (a false positive) is a **Type I error**.

---

> **9.** When a researcher fails to reject a false null hypothesis, a \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ error has been committed.

* **Answer:** **`Type II`**
* **Solution:**
  Failing to reject $H_0$ when $H_0$ is actually false (a false negative) is a **Type II error**.

---

> **10.** The probability of committing a Type II error is represented by \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`Beta`** ($\beta$)
* **Solution:**
  $\beta = P(\text{Fail to Reject } H_0 \mid H_0 \text{ is false})$ denotes the probability of a Type II error.

---

> **11.** Power is equal to \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`1 - β`**
* **Solution:**
  Statistical power is the probability of correctly rejecting a false null hypothesis: $\text{Power} = 1 - \beta$.

---

> **12.** Whenever hypotheses are established such that the alternative hypothesis is directional, then the researcher is conducting a \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_-tailed test.

* **Answer:** **`One`**
* **Solution:**
  Directional alternative hypotheses ($<$ or $>$) place the entire rejection region $\alpha$ in a single tail, producing a **one-tailed** test.

---

> **13.** A \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_-tailed test is nondirectional.

* **Answer:** **`Two`**
* **Solution:**
  A nondirectional alternative hypothesis ($\neq$) splits $\alpha$ equally across both tails ($\alpha / 2$ per tail), resulting in a **two-tailed** test.

---

> **14.** If in testing hypotheses, the researcher uses a method in which the probability of the observed statistic is compared to alpha to reach a decision, the researcher is using the \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ method.

* **Answer:** **`p-value`**
* **Solution:**
  In the **$p$-value method**, $H_0$ is rejected whenever the observed $p$-value is less than $\alpha$.

## Hypothesis tests for a population mean ($\sigma$ known, $z$-test) — Questions 15–24

> **15.** Suppose $H_0: \mu = 95$ and $H_a: \mu \neq 95$. If the sample size is $50$, the population standard deviation is known, and $\alpha = .05$, the critical value of $z$ is \_\_\_\_\_\_\_\_.

* **Answer:** **`±1.96`**
* **Solution:**
  Because $H_a: \mu \neq 95$ is two-tailed, $\alpha / 2 = 0.025$ is placed in each tail. From the standard normal distribution, $z_{\text{crit}} = \pm 1.96$.

---

> **16.** Suppose $H_0: \mu = 2.36$ and $H_a: \mu < 2.36$. If the sample size is $64$, the population standard deviation is known, and $\alpha = .01$, the critical value of $z$ is \_\_\_\_\_\_\_\_.

* **Answer:** **`-2.33`**
* **Solution:**
  Because $H_a: \mu < 2.36$ is left-tailed with $\alpha = 0.01$, the critical value in the lower tail is $z_{\text{crit}} = -2.33$ (exact `qnorm(0.01) = -2.3263`).

---

> **17.** Suppose $H_0: \mu = 24.8$ and $H_a: \mu \neq 24.8$. If the sample size is $49$, the population standard deviation is known, and $\alpha = .10$, the critical value of $z$ is \_\_\_\_\_\_\_\_.

* **Answer:** **`±1.645`**
* **Solution:**
  For a two-tailed test with $\alpha = 0.10$ ($\alpha / 2 = 0.05$ per tail), $z_{\text{crit}} = \pm 1.645$.

---

> **18.** Suppose a researcher is testing a null hypothesis that $\mu = 61$. A random sample of $n = 38$ is taken resulting in $\bar{x} = 63$ and $\sigma = 8.76$. The observed $z$ value is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`1.41`**
* **Solution:**
  $$z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}} = \frac{63 - 61}{8.76 / \sqrt{38}} = \frac{2}{1.42107} \approx 1.41$$

---

> **19.** Suppose a researcher is testing a null hypothesis that $\mu = 413$. A random sample of $n = 70$ is taken resulting in $\bar{x} = 405$. The population standard deviation is $34$. The observed $z$ value is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`-1.97`**
* **Solution:**
  $$z = \frac{405 - 413}{34 / \sqrt{70}} = \frac{-8}{4.06378} \approx -1.97$$

---

> **20.** A researcher is testing a hypothesis of a single mean. The critical $z$ value for $\alpha = .05$ and a one-tailed test is $1.645$. The observed $z$ value from sample data is $1.13$. The decision made by the researcher based on this information is to \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ the null hypothesis.

> **21.** A researcher is testing a hypothesis of a single mean. The critical $z$ value for $\alpha = .05$ and a two-tailed test is $\pm 1.96$. The observed $z$ value from sample data is $-1.91$. > **22.** A researcher is testing a hypothesis of a single mean. The critical $z$ value for $\alpha = .01$ and a one-tailed test is $-2.33$. The observed $z$ value from sample data is $-2.45$. The decision made by the researcher based on this information is to \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ the null hypothesis.

> **23.** A researcher has a theory that the average age of managers in a particular industry is over $35$-years-old. The null hypothesis to conduct a statistical test on this theory would be \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

## Hypothesis tests for a population proportion ($p$) — Questions 31–38

> **31.** A political scientist wants to statistically test the null hypothesis that her candidate for governor is currently carrying at least $57\%$ of the vote in the state. She has her assistants randomly sample $550$ eligible voters in the state by telephone and only $300$ declare that they support her candidate. The observed $z$ value for this problem is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`-1.16`**
* **Solution:**
  With $H_0: p \ge 0.57$ ($p_0 = 0.57$), $H_a: p < 0.57$, and $\hat{p} = \frac{300}{550} \approx 0.54545$:
  $$z = \frac{\hat{p} - p_0}{\sqrt{\frac{p_0(1 - p_0)}{n}}} = \frac{0.54545 - 0.57}{\sqrt{\frac{0.57(0.43)}{550}}} = \frac{-0.02455}{0.02111} \approx -1.16$$

---

> **32.** Problem 31 is a \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_-tailed test.

* **Answer:** **`One`** (specifically, left-tailed)
* **Solution:**
  Testing "at least $57\%$" ($H_0: p \ge 0.57$) against $H_a: p < 0.57$ places the entire rejection region in the lower tail, making it a **one-tailed** test.

---

> **33.** Suppose that the value of alpha for problem 31 is $.05$. After comparing the observed value to the critical value, the political scientist decided to \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ the null hypothesis.

* **Answer:** **`20.0902`**

## Type II error ($\beta$), statistical power ($1 - \beta$), and $p$-values — Questions 43–50

> **43.** The null hypothesis for a test is $H_0: \mu = 30$. The population standard deviation is known to be $0.63$. A one-tailed test is being conducted in the lower tail of the distribution. After taking a sample of $49$ items and computing a mean, it is decided to fail to reject the null hypothesis. Let $\alpha = .05$. If the null hypothesis is not true and if the true alternative hypothesis is $29.6$, the value of beta is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`.0026`**
* **Solution:**
  1. Critical sample mean under $H_0$ ($\alpha = 0.05$, lower tail, $z_{\alpha} = -1.645$):
     $$\bar{x}_c = 30 - 1.645\left(\frac{0.63}{\sqrt{49}}\right) = 30 - 1.645(0.09) = 29.85195$$
  2. Standardize $\bar{x}_c$ under $\mu_a = 29.6$:
     $$z_{\beta} = \frac{29.85195 - 29.6}{0.09} = \frac{0.25195}{0.09} \approx 2.80$$
  3. Since we fail to reject $H_0$ when $\bar{x} \ge \bar{x}_c$:
     $$\beta = P(Z \ge 2.80) = 1 - 0.9974 = 0.0026$$

---

> **44.** Suppose the alternative mean in problem 43 is really $29.9$, the value of beta is \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_.

* **Answer:** **`.7019`** (or `0.7033` using unrounded `qnorm(0.05)`)
* **Solution:**
  Using $\bar{x}_c = 29.85195$ and $\mu_a = 29.9$:
  $$z_{\beta} = \frac{29.85195 - 29.9}{0.09} = \frac{-0.04805}{0.09} \approx -0.53$$
  $$\beta = P(Z \ge -0.53) = 0.7019$$

---

> **45.** Plotting the power values against the various values of the alternative hypotheses produces a \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ curve.

* **Answer:** **`Power`**
* **Solution:**

> **46.** Plotting the values of $\beta$ against various values of the alternative hypothesis produces a \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ curve.

* **Answer:** **`Operating Characteristic`** (OC)
* **Solution:**
  An **Operating Characteristic (OC) Curve** plots $\beta$ (Type II error probability) across candidate alternative values $\mu_a$.

---

> **47.** The $p$-value for an observed $z$ of $2.73$ is \_\_\_\_\_\_\_\_\_.

> **48.** The $p$-value for an observed $z$ of $0.85$ is \_\_\_\_\_\_\_\_\_.

* **Answer:** **`.1977`**

---

> **49.** In a hypothesis-testing problem, a $p$-value of $.0046$ is obtained for the observed statistic. If a one-tailed test is being conducted and $\alpha$ is $.01$, then the decision is to \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ the null hypothesis.

* **Answer:** **`Reject`**

> **50.** A researcher is conducting a two-tailed hypothesis test using a $5\%$ level of significance. As a result of the test, a $p$-value of $.032$ is obtained for one of the tails. The decision should be to \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ the null hypothesis.

* **Answer:** **`Fail to Reject`**
* **Solution:**
  In a two-tailed test, the overall $p$-value is twice the single-tail probability: $p\text{-value} = 2(0.032) = 0.064 > \alpha = 0.05$ (equivalently, $0.032 > \alpha / 2 = 0.025$). Therefore, **Fail to Reject** $H_0$.

## Parametric vs. nonparametric statistics & goodness-of-fit tests — Questions 1–11

> **1.** Statistical techniques based on assumptions about the population from which the sample data are selected are called \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ statistics.

* **Answer:** **`Parametric`**
* **Solution:**
  **Parametric statistics** rely on explicit distributional assumptions (such as normality) and estimate or test population parameters like $\mu$ or $\sigma^2$.

---

> **2.** Statistical techniques based on fewer assumptions about the population and the parameters are called \_\_\_\_\_\_\_\_\_\_\_\_\_\_\_ statistics.

* **Answer:** **`Nonparametric`**
* **Solution:**

```r
print(sprintf("Q4 crit_chi2 (df=6, a=0.05): %.4f | Q6 crit_chi2 (df=4, a=0.01): %.4f",
              qchisq(0.95, df = 6), qchisq(0.99, df = 4)))

obs5 <- c(27, 30, 29, 21, 24)
print(sprintf("Q7 chi2: %.3f", chisq.test(obs5)$statistic))

k8 <- 0:5
obs8 <- c(47, 56, 38, 23, 15, 12)
n8 <- sum(obs8)
lambda_hat <- sum(k8 * obs8) / n8
probs8 <- c(dpois(0:4, lambda_hat), 1 - ppois(4, lambda_hat))
exp8 <- n8 * probs8
chi8 <- sum((obs8 - exp8)^2 / exp8)
print(sprintf("Q8 crit_chi2 (df=4, a=0.10): %.5f | Q9 lambda_hat: %.4f | chi2: %.2f",
              qchisq(0.90, df = 4), lambda_hat, chi8))

mat14 <- matrix(c(25, 40, 60, 10, 15, 20), nrow = 2, byrow = TRUE)
res14 <- chisq.test(mat14)
print(sprintf("Q14 E(D,B): %.2f | Q16 crit_chi2: %.4f | Q17 chi2: %.4f",
              res14$expected[1, 2], qchisq(0.95, df = 2), res14$statistic))

mat18 <- matrix(c(92, 70, 112, 145), nrow = 2, byrow = TRUE)
res18 <- chisq.test(mat18, correct = FALSE)
print(sprintf("Q18 crit_chi2: %.4f | Q19 chi2 (uncorrected): %.3f",
              qchisq(0.99, df = 1), res18$statistic))
```

### Exercise 3: Variance of beef hot dog sodium (`hot_dogs.csv`)

Using `hot_dogs` data and a chi-square hypothesis test (and corresponding $95\%$ confidence interval for $\sigma^2$), determine if the variance in `Sodium` values for `Beef` hot dogs is different from $6000$ with $95\%$ confidence.

```r
var_conf_int <- function(x, conf_level = 0.95) {
  df <- length(x) - 1
  chi_lower <- qchisq((1 - conf_level) / 2, df, lower.tail = TRUE)
  chi_upper <- qchisq((1 - conf_level) / 2, df, lower.tail = FALSE)
  v <- var(x)
  c(df * v / chi_upper, df * v / chi_lower)
}

ci_beef_sod <- var_conf_int(beef$Sodium, conf_level = 0.95)
df_beef <- nrow(beef) - 1
chi2_beef <- df_beef * var(beef$Sodium) / 6000
p_beef_var <- 2 * min(pchisq(chi2_beef, df = df_beef, lower.tail = TRUE),
                      pchisq(chi2_beef, df = df_beef, lower.tail = FALSE))

print(sprintf("Beef Sodium s^2 = %.2f | chi^2 = %.4f (df = %d) | p-value = %.4f",
              var(beef$Sodium), chi2_beef, df_beef, p_beef_var))
print(sprintf("95%% CI for variance: [%.3f, %.3f] | 6000 outside CI: %s",
              ci_beef_sod[1], ci_beef_sod[2],
              (6000 < ci_beef_sod[1]) || (6000 > ci_beef_sod[2])))
```

### Exercise 7: Dental tooth filling expenditures (Demonstration Problem 9.1 — One-sample $z$-test)

The historical average out-of-pocket expenditure for a dental filling is $\$141$ with population standard deviation $\sigma = \$30$. Using a random sample of $n = 32$ recent fillings and $\alpha = 0.05$, test whether current expenditures exceed $\$141$ ($H_a: \mu > 141$).

```r
fillings <- c(147, 145, 175, 163, 152, 166, 125, 175, 119, 149, 147, 120, 142,
              130, 210, 135, 163, 158, 132, 207, 151, 175, 156, 141, 123, 142,
              139, 170, 118, 160, 122, 110)

greater_than_z_function <- function(data, hist_mu, pop_sd, alpha) {
  sample_mu <- mean(x = data, na.rm = TRUE)
  z <- (sample_mu - hist_mu) / (pop_sd / sqrt(length(data)))
  crit_z <- qnorm(p = 1 - alpha, mean = 0, sd = 1, lower.tail = TRUE)
  outcome <- ifelse(z > crit_z, "Reject", "Fail to Reject")
  p_value <- pnorm(q = z, mean = 0, sd = 1, lower.tail = FALSE)
  list(sample_mean = sample_mu, z = z, crit_z = crit_z,
       null_hypothesis = outcome, p_value = p_value)
}

print(greater_than_z_function(fillings, hist_mu = 141, pop_sd = 30, alpha = 0.05))
```
